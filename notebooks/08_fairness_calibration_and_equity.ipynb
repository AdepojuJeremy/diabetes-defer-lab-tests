```python
def save_json(
    obj,
    path
):
    path = Path(path)

    path.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    with open(
        path,
        "w"
    ) as f:
        json.dump(
            obj,
            f,
            indent=2,
            default=str
        )

    return path


def save_plotly(
    fig,
    name,
    width=1200,
    height=750
):
    html_path = (
        FIGURE_DIR
        / f"{name}.html"
    )

    png_path = (
        FIGURE_DIR
        / f"{name}.png"
    )

    fig.write_html(
        html_path
    )

    try:
        fig.write_image(
            png_path,
            width=width,
            height=height,
            scale=2
        )

    except Exception as error:
        print(
            f"PNG export skipped for "
            f"{name}: {error}"
        )

        png_path = None

    return {
        "html": html_path,
        "png": png_path
    }


def safe_divide(
    numerator,
    denominator
):
    if denominator == 0:
        return np.nan

    return (
        numerator
        / denominator
    )
```

```python
cohort_path = (
    PROCESSED_DATA_DIR
    / "step7_primary_modeling_cohort.parquet"
)

assignment_path = (
    PROCESSED_DATA_DIR
    / "step11_budget_assignments.parquet"
)

feature_metadata_path = (
    REPORT_DIR
    / "step7_feature_metadata.json"
)

checkpoint_path = (
    STEP10_MODEL_DIR
    / "neural_defer_to_lab_final.pt"
)

for path in [
    cohort_path,
    assignment_path,
    feature_metadata_path,
    checkpoint_path
]:
    if not path.exists():
        raise FileNotFoundError(
            f"Missing required artifact: {path}"
        )

cohort = pd.read_parquet(
    cohort_path,
    engine="fastparquet"
)

assignments = pd.read_parquet(
    assignment_path,
    engine="fastparquet"
)

with open(
    feature_metadata_path,
    "r"
) as f:
    metadata = json.load(f)

TARGET_COL = metadata["target"]

test_cohort = (
    cohort[
        cohort["split"] == "test"
    ]
    .copy()
)

print(
    "Test cohort:",
    test_cohort.shape
)

print(
    "Budget assignments:",
    assignments.shape
)
```

```python
MIN_GROUP_N = 100
MIN_POSITIVE_N = 20
MIN_NEGATIVE_N = 20


def compute_group_metrics(
    df,
    group_variable,
    target_col
):
    records = []

    for group_name, group_df in df.groupby(
        group_variable,
        observed=True,
        dropna=False
    ):

        y = (
            group_df[
                target_col
            ]
            .astype(int)
            .to_numpy()
        )

        final_prediction = (
            group_df[
                "cascade_prediction"
            ]
            .astype(int)
            .to_numpy()
        )

        referral = (
            group_df[
                "referred"
            ]
            .astype(int)
            .to_numpy()
        )

        cheap_probability = (
            group_df[
                "cheap_probability"
            ]
            .astype(float)
            .to_numpy()
        )

        cascade_probability = (
            group_df[
                "cascade_probability"
            ]
            .astype(float)
            .to_numpy()
        )

        n = len(
            group_df
        )

        positives = int(
            y.sum()
        )

        negatives = int(
            n - positives
        )

        if n < MIN_GROUP_N:
            continue

        tn, fp, fn, tp = (
            confusion_matrix(
                y,
                final_prediction,
                labels=[0, 1]
            )
            .ravel()
        )

        sensitivity = safe_divide(
            tp,
            tp + fn
        )

        specificity = safe_divide(
            tn,
            tn + fp
        )

        ppv = safe_divide(
            tp,
            tp + fp
        )

        npv = safe_divide(
            tn,
            tn + fn
        )

        referral_rate = float(
            referral.mean()
        )

        positive_referral_rate = (
            float(
                referral[
                    y == 1
                ].mean()
            )
            if positives > 0
            else np.nan
        )

        negative_referral_rate = (
            float(
                referral[
                    y == 0
                ].mean()
            )
            if negatives > 0
            else np.nan
        )

        under_referral_cases = (
            1.0
            - positive_referral_rate
            if pd.notna(
                positive_referral_rate
            )
            else np.nan
        )

        referral_yield = (
            float(
                y[
                    referral == 1
                ].mean()
            )
            if referral.sum() > 0
            else np.nan
        )

        if (
            positives >= MIN_POSITIVE_N
            and
            negatives >= MIN_NEGATIVE_N
        ):
            cheap_auroc = roc_auc_score(
                y,
                cheap_probability
            )

            cheap_auprc = (
                average_precision_score(
                    y,
                    cheap_probability
                )
            )

            cheap_brier = (
                brier_score_loss(
                    y,
                    cheap_probability
                )
            )

            cascade_brier = (
                brier_score_loss(
                    y,
                    cascade_probability
                )
            )

        else:
            cheap_auroc = np.nan
            cheap_auprc = np.nan
            cheap_brier = np.nan
            cascade_brier = np.nan

        records.append({
            "group_variable": (
                group_variable
            ),

            "group": str(
                group_name
            ),

            "n": int(
                n
            ),

            "positives": (
                positives
            ),

            "negatives": (
                negatives
            ),

            "prevalence": float(
                y.mean()
            ),

            "referral_rate": (
                referral_rate
            ),

            "positive_case_referral_rate": (
                positive_referral_rate
            ),

            "negative_case_referral_rate": (
                negative_referral_rate
            ),

            "under_referral_rate_cases": (
                under_referral_cases
            ),

            "referral_yield": (
                referral_yield
            ),

            "accuracy": (
                accuracy_score(
                    y,
                    final_prediction
                )
            ),

            "balanced_accuracy": (
                balanced_accuracy_score(
                    y,
                    final_prediction
                )
            ),

            "sensitivity": (
                sensitivity
            ),

            "specificity": (
                specificity
            ),

            "ppv": (
                ppv
            ),

            "npv": (
                npv
            ),

            "f1": (
                f1_score(
                    y,
                    final_prediction,
                    zero_division=0
                )
            ),

            "false_negative_rate": (
                1.0 - sensitivity
                if pd.notna(
                    sensitivity
                )
                else np.nan
            ),

            "false_positive_rate": (
                1.0 - specificity
                if pd.notna(
                    specificity
                )
                else np.nan
            ),

            "cheap_auroc": (
                cheap_auroc
            ),

            "cheap_auprc": (
                cheap_auprc
            ),

            "cheap_brier": (
                cheap_brier
            ),

            "cascade_brier": (
                cascade_brier
            )
        })

    return pd.DataFrame(
        records
    )
```

```python
subgroup_metric_tables = []

for subgroup in SUBGROUP_VARIABLES:

    subgroup_result = (
        compute_group_metrics(
            analysis_df,
            subgroup,
            TARGET_COL
        )
    )

    if subgroup_result.empty:
        continue

    subgroup_metric_tables.append(
        subgroup_result
    )

    subgroup_result.to_csv(
        TABLE_DIR
        / f"step12_metrics_by_{subgroup}.csv",
        index=False
    )


subgroup_metrics = pd.concat(
    subgroup_metric_tables,
    ignore_index=True
)

display(
    subgroup_metrics
)

subgroup_metrics.to_csv(
    TABLE_DIR
    / "step12_all_subgroup_metrics.csv",
    index=False
)
```

```python
equity_gap_records = []

for subgroup in (
    subgroup_metrics[
        "group_variable"
    ].unique()
):

    subset = (
        subgroup_metrics[
            subgroup_metrics[
                "group_variable"
            ]
            == subgroup
        ]
    )

    metrics_to_compare = [
        "referral_rate",
        "positive_case_referral_rate",
        "under_referral_rate_cases",
        "sensitivity",
        "specificity",
        "false_negative_rate",
        "false_positive_rate",
        "balanced_accuracy",
        "cheap_brier"
    ]

    for metric in metrics_to_compare:

        valid = (
            subset[
                metric
            ]
            .dropna()
        )

        if len(valid) < 2:
            continue

        max_index = (
            subset[
                metric
            ]
            .idxmax()
        )

        min_index = (
            subset[
                metric
            ]
            .idxmin()
        )

        equity_gap_records.append({
            "group_variable": (
                subgroup
            ),

            "metric": (
                metric
            ),

            "maximum": float(
                valid.max()
            ),

            "minimum": float(
                valid.min()
            ),

            "absolute_gap": float(
                valid.max()
                - valid.min()
            ),

            "highest_group": (
                subset.loc[
                    max_index,
                    "group"
                ]
            ),

            "lowest_group": (
                subset.loc[
                    min_index,
                    "group"
                ]
            )
        })


equity_gaps = pd.DataFrame(
    equity_gap_records
)

display(
    equity_gaps.sort_values(
        "absolute_gap",
        ascending=False
    )
)

equity_gaps.to_csv(
    TABLE_DIR
    / "step12_equity_gaps.csv",
    index=False
)
```

```python
def expected_calibration_error(
    y_true,
    probability,
    n_bins=10
):
    y_true = np.asarray(
        y_true
    ).astype(int)

    probability = np.asarray(
        probability,
        dtype=float
    )

    bins = np.linspace(
        0,
        1,
        n_bins + 1
    )

    bin_ids = np.digitize(
        probability,
        bins[1:-1],
        right=True
    )

    ece = 0.0

    for bin_id in range(
        n_bins
    ):

        mask = (
            bin_ids
            == bin_id
        )

        if mask.sum() == 0:
            continue

        observed = (
            y_true[
                mask
            ].mean()
        )

        predicted = (
            probability[
                mask
            ].mean()
        )

        weight = (
            mask.mean()
        )

        ece += (
            weight
            * abs(
                observed
                - predicted
            )
        )

    return float(
        ece
    )
```

```python
calibration_metric_records = []

for subgroup in SUBGROUP_VARIABLES:

    for group_name, group_df in analysis_df.groupby(
        subgroup,
        observed=True,
        dropna=False
    ):

        if len(
            group_df
        ) < MIN_GROUP_N:
            continue

        y = (
            group_df[
                TARGET_COL
            ]
            .astype(int)
            .to_numpy()
        )

        probability = (
            group_df[
                "cheap_probability"
            ]
            .astype(float)
            .to_numpy()
        )

        if (
            y.sum()
            < MIN_POSITIVE_N
            or
            (len(y) - y.sum())
            < MIN_NEGATIVE_N
        ):
            continue

        calibration_metric_records.append({
            "group_variable": (
                subgroup
            ),

            "group": str(
                group_name
            ),

            "n": int(
                len(
                    group_df
                )
            ),

            "prevalence": float(
                y.mean()
            ),

            "mean_predicted_risk": float(
                probability.mean()
            ),

            "brier_score": (
                brier_score_loss(
                    y,
                    probability
                )
            ),

            "ece_10_bins": (
                expected_calibration_error(
                    y,
                    probability,
                    n_bins=10
                )
            )
        })


calibration_metrics = pd.DataFrame(
    calibration_metric_records
)

display(
    calibration_metrics.sort_values(
        "ece_10_bins",
        ascending=False
    )
)

calibration_metrics.to_csv(
    TABLE_DIR
    / "step12_subgroup_calibration_metrics.csv",
    index=False
)
```

```python
def build_calibration_curve_table(
    df,
    group_variable,
    target_col,
    probability_col,
    n_bins=8
):
    records = []

    for group_name, group_df in df.groupby(
        group_variable,
        observed=True,
        dropna=False
    ):

        if len(
            group_df
        ) < MIN_GROUP_N:
            continue

        work = (
            group_df[
                [
                    target_col,
                    probability_col
                ]
            ]
            .dropna()
            .copy()
        )

        if len(
            work
        ) < MIN_GROUP_N:
            continue

        work[
            "calibration_bin"
        ] = pd.qcut(
            work[
                probability_col
            ].astype(
                "float32"
            ),
            q=min(
                n_bins,
                work[
                    probability_col
                ].nunique()
            ),
            labels=False,
            duplicates="drop"
        )

        summary = (
            work.groupby(
                "calibration_bin",
                observed=True
            )
            .agg(
                mean_predicted=(
                    probability_col,
                    "mean"
                ),

                observed_rate=(
                    target_col,
                    "mean"
                ),

                n=(
                    target_col,
                    "size"
                )
            )
            .reset_index()
        )

        summary[
            "group"
        ] = str(
            group_name
        )

        summary[
            "group_variable"
        ] = (
            group_variable
        )

        records.append(
            summary
        )

    if not records:
        return pd.DataFrame()

    return pd.concat(
        records,
        ignore_index=True
    )
```

```python
BOOTSTRAP_REPLICATES = 250
BOOTSTRAP_SEED = 42

rng = np.random.default_rng(
    BOOTSTRAP_SEED
)


def bootstrap_group_metrics(
    group_df,
    target_col,
    n_bootstrap=250
):
    estimates = []

    n = len(
        group_df
    )

    if n < MIN_GROUP_N:
        return None

    for _ in range(
        n_bootstrap
    ):

        indices = rng.integers(
            0,
            n,
            size=n
        )

        sample = (
            group_df.iloc[
                indices
            ]
        )

        y = (
            sample[
                target_col
            ]
            .astype(int)
            .to_numpy()
        )

        prediction = (
            sample[
                "cascade_prediction"
            ]
            .astype(int)
            .to_numpy()
        )

        referral = (
            sample[
                "referred"
            ]
            .astype(int)
            .to_numpy()
        )

        positives = (
            y == 1
        )

        negatives = (
            y == 0
        )

        if (
            positives.sum() == 0
            or
            negatives.sum() == 0
        ):
            continue

        tn, fp, fn, tp = (
            confusion_matrix(
                y,
                prediction,
                labels=[0, 1]
            )
            .ravel()
        )

        estimates.append({
            "referral_rate": (
                referral.mean()
            ),

            "positive_case_referral_rate": (
                referral[
                    positives
                ].mean()
            ),

            "sensitivity": safe_divide(
                tp,
                tp + fn
            ),

            "specificity": safe_divide(
                tn,
                tn + fp
            )
        })

    if not estimates:
        return None

    bootstrap_df = (
        pd.DataFrame(
            estimates
        )
    )

    results = {}

    for metric in (
        bootstrap_df.columns
    ):

        results[
            f"{metric}_low"
        ] = float(
            bootstrap_df[
                metric
            ].quantile(
                0.025
            )
        )

        results[
            f"{metric}_high"
        ] = float(
            bootstrap_df[
                metric
            ].quantile(
                0.975
            )
        )

    return results
```

```python
bootstrap_records = []

for subgroup in SUBGROUP_VARIABLES:

    for group_name, group_df in (
        analysis_df.groupby(
            subgroup,
            observed=True,
            dropna=False
        )
    ):

        if len(
            group_df
        ) < MIN_GROUP_N:
            continue

        result = (
            bootstrap_group_metrics(
                group_df,
                TARGET_COL,
                n_bootstrap=(
                    BOOTSTRAP_REPLICATES
                )
            )
        )

        if result is None:
            continue

        bootstrap_records.append({
            "group_variable": (
                subgroup
            ),

            "group": str(
                group_name
            ),

            "n": int(
                len(
                    group_df
                )
            ),

            **result
        })


bootstrap_intervals = (
    pd.DataFrame(
        bootstrap_records
    )
)

display(
    bootstrap_intervals
)

bootstrap_intervals.to_csv(
    TABLE_DIR
    / "step12_subgroup_bootstrap_intervals.csv",
    index=False
)
```

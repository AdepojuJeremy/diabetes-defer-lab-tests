```python
def save_json(obj, path):
    path = Path(path)

    path.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    with open(path, "w") as f:
        json.dump(
            obj,
            f,
            indent=2,
            default=str
        )

    return path


def save_plotly(
    fig,
    name,
    width=1200,
    height=750
):
    html_path = (
        FIGURE_DIR
        / f"{name}.html"
    )

    png_path = (
        FIGURE_DIR
        / f"{name}.png"
    )

    fig.write_html(
        html_path
    )

    try:
        fig.write_image(
            png_path,
            width=width,
            height=height,
            scale=2
        )

        print(
            "PNG saved:",
            png_path
        )

    except Exception as error:
        print(
            f"PNG export skipped for {name}: {error}"
        )

        png_path = None

    return {
        "html": html_path,
        "png": png_path
    }
```

```python
import torch
import torch.nn as nn


class MLPBlock(nn.Module):

    def __init__(
        self,
        input_dim,
        hidden_dim=128,
        num_layers=3,
        dropout=0.15
    ):
        super().__init__()

        layers = []
        current_dim = input_dim

        for _ in range(num_layers):

            layers.extend([
                nn.Linear(
                    current_dim,
                    hidden_dim
                ),

                nn.LayerNorm(
                    hidden_dim
                ),

                nn.GELU(),

                nn.Dropout(
                    dropout
                )
            ])

            current_dim = hidden_dim

        self.network = nn.Sequential(
            *layers
        )

    def forward(self, x):
        return self.network(x)


class DeferToLabNet(nn.Module):

    def __init__(
        self,
        cheap_input_dim,
        full_input_dim,
        hidden_dim=128,
        num_layers=3,
        dropout=0.15
    ):
        super().__init__()

        self.cheap_encoder = MLPBlock(
            input_dim=cheap_input_dim,
            hidden_dim=hidden_dim,
            num_layers=num_layers,
            dropout=dropout
        )

        self.full_encoder = MLPBlock(
            input_dim=full_input_dim,
            hidden_dim=hidden_dim,
            num_layers=num_layers,
            dropout=dropout
        )

        self.gate_encoder = MLPBlock(
            input_dim=cheap_input_dim,
            hidden_dim=hidden_dim,
            num_layers=2,
            dropout=dropout
        )

        self.cheap_head = nn.Linear(
            hidden_dim,
            1
        )

        self.full_head = nn.Linear(
            hidden_dim,
            1
        )

        self.defer_head = nn.Linear(
            hidden_dim,
            1
        )

    def forward(
        self,
        x_cheap,
        x_full
    ):

        cheap_representation = (
            self.cheap_encoder(
                x_cheap
            )
        )

        full_representation = (
            self.full_encoder(
                x_full
            )
        )

        gate_representation = (
            self.gate_encoder(
                x_cheap
            )
        )

        cheap_logit = (
            self.cheap_head(
                cheap_representation
            )
            .squeeze(-1)
        )

        full_logit = (
            self.full_head(
                full_representation
            )
            .squeeze(-1)
        )

        defer_logit = (
            self.defer_head(
                gate_representation
            )
            .squeeze(-1)
        )

        return {
            "cheap_logit": cheap_logit,
            "full_logit": full_logit,
            "defer_logit": defer_logit,
            "defer_probability": torch.sigmoid(
                defer_logit
            )
        }


print("DeferToLabNet definition restored.")
```

```python
def map_transformed_to_original(
    transformed_name,
    original_features
):
    name = str(
        transformed_name
    )

    if "__" in name:
        name = name.split(
            "__",
            1
        )[1]

    if name.startswith(
        "missingindicator_"
    ):
        name = name.replace(
            "missingindicator_",
            "",
            1
        )

    candidate_features = sorted(
        original_features,
        key=len,
        reverse=True
    )

    for feature in candidate_features:

        if (
            name == feature
            or
            name.startswith(
                feature + "_"
            )
        ):
            return feature

    return name


transformed_to_original = [
    map_transformed_to_original(
        name,
        cheap_features
    )
    for name in transformed_feature_names
]

feature_mapping = pd.DataFrame({
    "transformed_feature": (
        transformed_feature_names
    ),

    "original_feature": (
        transformed_to_original
    )
})

display(
    feature_mapping.head(30)
)

feature_mapping.to_csv(
    TABLE_DIR
    / "step13_transformed_feature_mapping.csv",
    index=False
)
```

```python
class CheapLogitWrapper(nn.Module):

    def __init__(self, model):
        super().__init__()

        self.model = model

    def forward(self, x):

        representation = (
            self.model
            .cheap_encoder(
                x
            )
        )

        return (
            self.model
            .cheap_head(
                representation
            )
            .squeeze(-1)
        )


class DeferLogitWrapper(nn.Module):

    def __init__(self, model):
        super().__init__()

        self.model = model

    def forward(self, x):

        representation = (
            self.model
            .gate_encoder(
                x
            )
        )

        return (
            self.model
            .defer_head(
                representation
            )
            .squeeze(-1)
        )


cheap_wrapper = (
    CheapLogitWrapper(
        model
    )
    .to(DEVICE)
    .eval()
)

gate_wrapper = (
    DeferLogitWrapper(
        model
    )
    .to(DEVICE)
    .eval()
)

print(
    "Interpretability wrappers ready."
)
```

```python
def integrated_gradients(
    wrapper,
    X,
    baseline,
    n_steps=24,
    batch_size=128
):
    wrapper.eval()

    X = np.asarray(
        X,
        dtype=np.float32
    )

    baseline = np.asarray(
        baseline,
        dtype=np.float32
    )

    attributions = np.zeros_like(
        X,
        dtype=np.float32
    )

    alphas = np.linspace(
        0.0,
        1.0,
        n_steps,
        dtype=np.float32
    )

    for batch_start in range(
        0,
        len(X),
        batch_size
    ):

        batch_end = min(
            batch_start
            + batch_size,
            len(X)
        )

        x_batch = torch.tensor(
            X[
                batch_start:
                batch_end
            ],
            dtype=torch.float32,
            device=DEVICE
        )

        baseline_batch = torch.tensor(
            baseline,
            dtype=torch.float32,
            device=DEVICE
        ).unsqueeze(0).expand_as(
            x_batch
        )

        delta = (
            x_batch
            - baseline_batch
        )

        gradient_integral = (
            torch.zeros_like(
                x_batch
            )
        )

        for step_index, alpha in enumerate(
            alphas
        ):

            interpolated = (
                baseline_batch
                + float(alpha)
                * delta
            )

            interpolated.requires_grad_(
                True
            )

            output = wrapper(
                interpolated
            )

            gradients = torch.autograd.grad(
                outputs=output.sum(),
                inputs=interpolated,
                retain_graph=False,
                create_graph=False
            )[0]

            weight = 1.0

            if (
                step_index == 0
                or
                step_index
                == len(alphas) - 1
            ):
                weight = 0.5

            gradient_integral += (
                weight
                * gradients.detach()
            )

        gradient_integral /= (
            len(alphas)
            - 1
        )

        batch_attribution = (
            delta
            * gradient_integral
        )

        attributions[
            batch_start:
            batch_end
        ] = (
            batch_attribution
            .detach()
            .cpu()
            .numpy()
            .astype(
                np.float32
            )
        )

    return attributions
```

```python
IG_N_STEPS = 24

print(
    "Computing cheap-model Integrated Gradients..."
)

cheap_ig = integrated_gradients(
    wrapper=cheap_wrapper,
    X=X_interpret,
    baseline=IG_BASELINE,
    n_steps=IG_N_STEPS,
    batch_size=128
)

print(
    "Computing defer-gate Integrated Gradients..."
)

gate_ig = integrated_gradients(
    wrapper=gate_wrapper,
    X=X_interpret,
    baseline=IG_BASELINE,
    n_steps=IG_N_STEPS,
    batch_size=128
)

print(
    "Cheap IG:",
    cheap_ig.shape
)

print(
    "Gate IG:",
    gate_ig.shape
)
```

```python
def aggregate_attributions(
    attribution_matrix,
    feature_mapping,
    original_features
):
    feature_to_index = {
        feature: index
        for index, feature
        in enumerate(
            original_features
        )
    }

    aggregated = np.zeros(
        (
            attribution_matrix.shape[0],
            len(original_features)
        ),
        dtype=np.float32
    )

    for transformed_index, original_feature in enumerate(
        feature_mapping
    ):

        if (
            original_feature
            in feature_to_index
        ):

            aggregated[
                :,
                feature_to_index[
                    original_feature
                ]
            ] += (
                attribution_matrix[
                    :,
                    transformed_index
                ]
            )

    return aggregated


cheap_ig_original = (
    aggregate_attributions(
        cheap_ig,
        transformed_to_original,
        cheap_features
    )
)

gate_ig_original = (
    aggregate_attributions(
        gate_ig,
        transformed_to_original,
        cheap_features
    )
)

print(
    "Aggregated cheap IG:",
    cheap_ig_original.shape
)

print(
    "Aggregated gate IG:",
    gate_ig_original.shape
)
```

```python
importance_records = []

for feature_index, feature in enumerate(
    cheap_features
):

    importance_records.append({
        "feature": feature,

        "cheap_mean_abs_ig": float(
            np.mean(
                np.abs(
                    cheap_ig_original[
                        :,
                        feature_index
                    ]
                )
            )
        ),

        "cheap_mean_signed_ig": float(
            np.mean(
                cheap_ig_original[
                    :,
                    feature_index
                ]
            )
        ),

        "gate_mean_abs_ig": float(
            np.mean(
                np.abs(
                    gate_ig_original[
                        :,
                        feature_index
                    ]
                )
            )
        ),

        "gate_mean_signed_ig": float(
            np.mean(
                gate_ig_original[
                    :,
                    feature_index
                ]
            )
        )
    })


global_importance = (
    pd.DataFrame(
        importance_records
    )
)

global_importance[
    "cheap_importance_share"
] = (
    global_importance[
        "cheap_mean_abs_ig"
    ]
    /
    global_importance[
        "cheap_mean_abs_ig"
    ].sum()
)

global_importance[
    "gate_importance_share"
] = (
    global_importance[
        "gate_mean_abs_ig"
    ]
    /
    global_importance[
        "gate_mean_abs_ig"
    ].sum()
)

global_importance[
    "gate_minus_cheap_share"
] = (
    global_importance[
        "gate_importance_share"
    ]
    -
    global_importance[
        "cheap_importance_share"
    ]
)

display(
    global_importance
    .sort_values(
        "gate_mean_abs_ig",
        ascending=False
    )
)

global_importance.to_csv(
    TABLE_DIR
    / "step13_global_integrated_gradients.csv",
    index=False
)
```

```python
@torch.no_grad()
def predict_wrapper_probability(
    wrapper,
    X,
    batch_size=2048
):
    probabilities = []

    for start in range(
        0,
        len(X),
        batch_size
    ):

        end = min(
            start
            + batch_size,
            len(X)
        )

        x = torch.tensor(
            X[
                start:end
            ],
            dtype=torch.float32,
            device=DEVICE
        )

        logits = wrapper(x)

        probability = (
            torch.sigmoid(
                logits
            )
            .float()
            .cpu()
            .numpy()
        )

        probabilities.append(
            probability
        )

    return np.concatenate(
        probabilities
    ).astype(
        np.float32
    )


cheap_original_probability = (
    predict_wrapper_probability(
        cheap_wrapper,
        X_interpret
    )
)

gate_original_probability = (
    predict_wrapper_probability(
        gate_wrapper,
        X_interpret
    )
)
```

```python
ablation_records = []

for feature in cheap_features:

    transformed_indices = [
        index
        for index, mapped_feature
        in enumerate(
            transformed_to_original
        )
        if mapped_feature == feature
    ]

    if not transformed_indices:
        continue

    X_ablated = (
        X_interpret.copy()
    )

    X_ablated[
        :,
        transformed_indices
    ] = IG_BASELINE[
        transformed_indices
    ]

    cheap_ablated_probability = (
        predict_wrapper_probability(
            cheap_wrapper,
            X_ablated
        )
    )

    gate_ablated_probability = (
        predict_wrapper_probability(
            gate_wrapper,
            X_ablated
        )
    )

    cheap_delta = (
        cheap_original_probability
        -
        cheap_ablated_probability
    )

    gate_delta = (
        gate_original_probability
        -
        gate_ablated_probability
    )

    ablation_records.append({
        "feature": feature,

        "cheap_mean_abs_probability_change": float(
            np.abs(
                cheap_delta
            ).mean()
        ),

        "cheap_mean_signed_probability_change": float(
            cheap_delta.mean()
        ),

        "gate_mean_abs_probability_change": float(
            np.abs(
                gate_delta
            ).mean()
        ),

        "gate_mean_signed_probability_change": float(
            gate_delta.mean()
        )
    })


ablation_importance = pd.DataFrame(
    ablation_records
)

display(
    ablation_importance
    .sort_values(
        "gate_mean_abs_probability_change",
        ascending=False
    )
)

ablation_importance.to_csv(
    TABLE_DIR
    / "step13_feature_ablation_importance.csv",
    index=False
)
```

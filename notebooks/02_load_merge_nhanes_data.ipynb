```python
from pathlib import Path
from typing import Dict, List, Tuple, Optional, Any
import json
import time
import gc
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.graph_objects as go

BASE_DIR = Path.home() / "projects" / "diabetes-defer-lab-tests"

DATA_DIR = BASE_DIR / "data"
RAW_DATA_DIR = DATA_DIR / "raw_nhanes"
INTERIM_DATA_DIR = DATA_DIR / "interim"
PROCESSED_DATA_DIR = DATA_DIR / "processed"

OUTPUT_DIR = BASE_DIR / "outputs"
FIGURE_DIR = OUTPUT_DIR / "figures"
TABLE_DIR = OUTPUT_DIR / "tables"
MODEL_DIR = OUTPUT_DIR / "models"
LOG_DIR = OUTPUT_DIR / "logs"
REPORT_DIR = BASE_DIR / "reports"
CACHE_DIR = BASE_DIR / "cache"

for directory in [
    RAW_DATA_DIR,
    INTERIM_DATA_DIR,
    PROCESSED_DATA_DIR,
    FIGURE_DIR,
    TABLE_DIR,
    MODEL_DIR,
    LOG_DIR,
    REPORT_DIR,
    CACHE_DIR
]:
    directory.mkdir(parents=True, exist_ok=True)

PROJECT_COLORS = {
    "teal_dark": "#037688",
    "teal": "#4B9DAA",
    "teal_light": "#8CCBD3",
    "green_dark": "#488D6C",
    "green": "#7DAF97",
    "green_light": "#B7D6C6",
    "yellow_dark": "#C9A821",
    "yellow": "#EED371",
    "yellow_light": "#F6E6A6",
    "purple_dark": "#6E3C5D",
    "purple": "#9D7890",
    "purple_light": "#C1AFBF",
    "grey_dark": "#5F5F5D",
    "grey": "#8A8A87",
    "grey_light": "#D9D9D6",
    "black": "#222222",
    "white": "#FFFFFF",
    "background": "#FAFAF8"
}

class SimpleConfig:
    class data:
        min_age = 18

CONFIG = SimpleConfig()

def clean_column_names(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    df.columns = [str(col).strip() for col in df.columns]
    return df

def reduce_dataframe_memory(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()

    for column in df.select_dtypes(include=["float64"]).columns:
        df[column] = pd.to_numeric(df[column], downcast="float")

    for column in df.select_dtypes(include=["int64"]).columns:
        df[column] = pd.to_numeric(df[column], downcast="integer")

    return df

def memory_usage_report(df: pd.DataFrame, name: str = "dataframe") -> None:
    memory_mb = df.memory_usage(deep=True).sum() / 1024**2
    print(f"{name}: {df.shape[0]:,} rows x {df.shape[1]:,} columns, {memory_mb:.2f} MB")

def save_json(obj, path: Path) -> Path:
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)

    with open(path, "w") as f:
        json.dump(obj, f, indent=2, default=str)

    return path

def save_plotly_figure(fig, filename: str, width: int = 1200, height: int = 800):
    html_path = FIGURE_DIR / f"{filename}.html"
    png_path = FIGURE_DIR / f"{filename}.png"

    fig.write_html(html_path)

    try:
        fig.write_image(png_path, width=width, height=height, scale=2)
    except Exception as error:
        print(f"PNG export skipped for {filename}: {error}")
        png_path = None

    return {"html": html_path, "png": png_path}

print("Section 6 dependencies loaded.")
print("Raw files found:", len(list(RAW_DATA_DIR.glob("*.XPT"))))
```

```python
NHANES_CYCLE_SUFFIX_TO_YEARS = {
    "G": "2011-2012",
    "H": "2013-2014",
    "I": "2015-2016",
    "J": "2017-2018"
}

NHANES_MODULE_DESCRIPTIONS = {
    "DEMO": "Demographics",
    "BMX": "Body measures",
    "BPX": "Blood pressure examination",
    "BPQ": "Blood pressure questionnaire",
    "DIQ": "Diabetes questionnaire",
    "GHB": "Glycohemoglobin / HbA1c",
    "GLU": "Fasting glucose",
    "OGTT": "Oral glucose tolerance test",
    "SMQ": "Smoking questionnaire",
    "PAQ": "Physical activity questionnaire",
    "ALQ": "Alcohol questionnaire",
    "MCQ": "Medical conditions questionnaire"
}

CRITICAL_MODULES = ["DEMO", "BMX", "BPX", "DIQ", "GHB", "GLU"]
OPTIONAL_MODULES = ["BPQ", "OGTT", "SMQ", "PAQ", "ALQ", "MCQ"]


def infer_nhanes_module_and_suffix(path: Path) -> Dict[str, Any]:
    stem = path.stem.upper()
    parts = stem.split("_")

    if len(parts) == 1:
        module = parts[0]
        suffix = ""
    else:
        module = parts[0]
        suffix = parts[-1]

    cycle = NHANES_CYCLE_SUFFIX_TO_YEARS.get(suffix, "unknown")

    return {
        "filename": path.name,
        "path": str(path),
        "module": module,
        "suffix": suffix,
        "cycle": cycle,
        "description": NHANES_MODULE_DESCRIPTIONS.get(module, "Unknown module"),
        "is_critical": module in CRITICAL_MODULES,
        "is_optional": module in OPTIONAL_MODULES,
        "size_mb": round(path.stat().st_size / 1024**2, 3)
    }


def discover_nhanes_files(raw_data_dir: Path) -> pd.DataFrame:
    paths = sorted(list(Path(raw_data_dir).glob("*.XPT")) + list(Path(raw_data_dir).glob("*.xpt")))

    records = [infer_nhanes_module_and_suffix(path) for path in paths]

    registry = pd.DataFrame(records)

    if len(registry) == 0:
        raise FileNotFoundError(f"No NHANES .XPT files found in {raw_data_dir}")

    registry = registry.sort_values(["cycle", "module", "filename"]).reset_index(drop=True)

    return registry


file_registry = discover_nhanes_files(RAW_DATA_DIR)

registry_path = INTERIM_DATA_DIR / "step6_nhanes_file_registry.csv"
file_registry.to_csv(registry_path, index=False)

display(file_registry)

print(f"Files discovered: {len(file_registry)}")
print(f"Registry saved to: {registry_path}")
```

```python
coverage_table = (
    file_registry
    .pivot_table(
        index="cycle",
        columns="module",
        values="filename",
        aggfunc="count",
        fill_value=0
    )
    .reset_index()
)

display(coverage_table)

coverage_path = INTERIM_DATA_DIR / "step6_module_coverage_table.csv"
coverage_table.to_csv(coverage_path, index=False)

expected_records = []

for cycle in sorted(NHANES_CYCLE_SUFFIX_TO_YEARS.values()):
    for module in CRITICAL_MODULES:
        exists = (
            (file_registry["cycle"] == cycle)
            & (file_registry["module"] == module)
        ).any()

        expected_records.append({
            "cycle": cycle,
            "module": module,
            "required_level": "critical",
            "exists": bool(exists)
        })

    for module in OPTIONAL_MODULES:
        exists = (
            (file_registry["cycle"] == cycle)
            & (file_registry["module"] == module)
        ).any()

        expected_records.append({
            "cycle": cycle,
            "module": module,
            "required_level": "optional",
            "exists": bool(exists)
        })

expected_coverage = pd.DataFrame(expected_records)

missing_critical = expected_coverage[
    (expected_coverage["required_level"] == "critical")
    & (~expected_coverage["exists"])
].copy()

missing_optional = expected_coverage[
    (expected_coverage["required_level"] == "optional")
    & (~expected_coverage["exists"])
].copy()

display(expected_coverage)
display(missing_critical)
display(missing_optional)

expected_coverage.to_csv(INTERIM_DATA_DIR / "step6_expected_module_coverage.csv", index=False)
missing_critical.to_csv(INTERIM_DATA_DIR / "step6_missing_critical_modules.csv", index=False)
missing_optional.to_csv(INTERIM_DATA_DIR / "step6_missing_optional_modules.csv", index=False)

print(f"Missing critical modules: {len(missing_critical)}")
print(f"Missing optional modules: {len(missing_optional)}")

if len(missing_critical) > 0:
    raise RuntimeError("Critical NHANES files are missing. Stop and inspect missing_critical.")

print("Critical module coverage is sufficient.")
```

```python
def load_single_nhanes_file(path: Path) -> pd.DataFrame:
    df = pd.read_sas(
        path,
        format="xport",
        encoding="utf-8"
    )

    df = clean_column_names(df)

    if "SEQN" in df.columns:
        df["SEQN"] = pd.to_numeric(df["SEQN"], errors="coerce").astype("Int64")

    return df


def load_all_nhanes_files(file_registry: pd.DataFrame) -> Dict[Tuple[str, str], pd.DataFrame]:
    loaded = {}

    for row in file_registry.itertuples(index=False):
        path = Path(row.path)
        key = (row.cycle, row.module)

        try:
            df = load_single_nhanes_file(path)
            df["cycle"] = row.cycle
            df["cycle_suffix"] = row.suffix

            loaded[key] = df

            print(f"Loaded {row.filename}: {df.shape[0]:,} rows x {df.shape[1]:,} columns")

        except Exception as error:
            print(f"Failed to load {row.filename}: {error}")

    return loaded


loaded_nhanes = load_all_nhanes_files(file_registry)

print(f"Loaded module-cycle datasets: {len(loaded_nhanes)}")
```

```python
def create_variable_inventory(loaded_data: Dict[Tuple[str, str], pd.DataFrame]) -> pd.DataFrame:
    records = []

    for (cycle, module), df in loaded_data.items():
        for column in df.columns:
            records.append({
                "cycle": cycle,
                "module": module,
                "column": column,
                "dtype": str(df[column].dtype),
                "non_missing": int(df[column].notna().sum()),
                "missing": int(df[column].isna().sum()),
                "missing_rate": float(df[column].isna().mean()),
                "n_unique": int(df[column].nunique(dropna=True))
            })

    return pd.DataFrame(records)


variable_inventory = create_variable_inventory(loaded_nhanes)

variable_inventory_path = INTERIM_DATA_DIR / "step6_variable_inventory.csv"
variable_inventory.to_csv(variable_inventory_path, index=False)

display(variable_inventory.head(30))

print(f"Variable inventory rows: {len(variable_inventory):,}")
print(f"Saved to: {variable_inventory_path}")
```

```python
def merge_modules_for_cycle(
    loaded_data: Dict[Tuple[str, str], pd.DataFrame],
    cycle: str,
    module_order: List[str]
) -> pd.DataFrame:
    available_modules = [
        module for module in module_order
        if (cycle, module) in loaded_data
    ]

    if "DEMO" not in available_modules:
        raise ValueError(f"DEMO file is missing for cycle {cycle}")

    merged = loaded_data[(cycle, "DEMO")].copy()

    for module in available_modules:
        if module == "DEMO":
            continue

        right = loaded_data[(cycle, module)].copy()

        duplicate_columns = [
            col for col in right.columns
            if col in merged.columns and col not in ["SEQN"]
        ]

        right = right.drop(columns=duplicate_columns, errors="ignore")

        merged = merged.merge(
            right,
            on="SEQN",
            how="left",
            validate="one_to_one"
        )

    merged["cycle"] = cycle

    return merged


MODULE_MERGE_ORDER = [
    "DEMO",
    "BMX",
    "BPX",
    "BPQ",
    "DIQ",
    "GHB",
    "GLU",
    "OGTT",
    "SMQ",
    "PAQ",
    "ALQ",
    "MCQ"
]

available_cycles = sorted(file_registry["cycle"].dropna().unique().tolist())

cycle_panels = []

for cycle in available_cycles:
    if cycle == "unknown":
        continue

    panel = merge_modules_for_cycle(
        loaded_data=loaded_nhanes,
        cycle=cycle,
        module_order=MODULE_MERGE_ORDER
    )

    cycle_panels.append(panel)

    print(f"Merged {cycle}: {panel.shape[0]:,} rows x {panel.shape[1]:,} columns")

nhanes_merged_raw = pd.concat(cycle_panels, axis=0, ignore_index=True)

nhanes_merged_raw = reduce_dataframe_memory(nhanes_merged_raw)

memory_usage_report(nhanes_merged_raw, "nhanes_merged_raw")

merged_raw_path = INTERIM_DATA_DIR / "step6_nhanes_merged_raw.parquet"
nhanes_merged_raw.to_parquet(merged_raw_path, index=False)

print(f"Merged raw NHANES panel saved to: {merged_raw_path}")
```

```python
def get_first_existing_column(df: pd.DataFrame, candidates: List[str]) -> Optional[str]:
    for column in candidates:
        if column in df.columns:
            return column
    return None


def recode_sex(series: pd.Series) -> pd.Series:
    mapping = {
        1: "Male",
        2: "Female"
    }

    return pd.to_numeric(series, errors="coerce").map(mapping)


def recode_yes_no(series: pd.Series) -> pd.Series:
    numeric = pd.to_numeric(series, errors="coerce")

    return np.select(
        [
            numeric == 1,
            numeric == 2
        ],
        [
            1,
            0
        ],
        default=np.nan
    )


def make_age_group(age: pd.Series) -> pd.Series:
    age_numeric = pd.to_numeric(age, errors="coerce")

    return pd.cut(
        age_numeric,
        bins=[17, 34, 44, 54, 64, 74, 85],
        labels=["18-34", "35-44", "45-54", "55-64", "65-74", "75+"],
        include_lowest=True
    )


def make_bmi_group(bmi: pd.Series) -> pd.Series:
    bmi_numeric = pd.to_numeric(bmi, errors="coerce")

    return pd.cut(
        bmi_numeric,
        bins=[0, 18.5, 25, 30, 35, 40, 100],
        labels=[
            "Underweight",
            "Normal",
            "Overweight",
            "Obesity I",
            "Obesity II",
            "Obesity III"
        ],
        include_lowest=True
    )


def row_mean_existing(df: pd.DataFrame, columns: List[str]) -> pd.Series:
    existing = [col for col in columns if col in df.columns]

    if len(existing) == 0:
        return pd.Series(np.nan, index=df.index)

    return df[existing].apply(pd.to_numeric, errors="coerce").mean(axis=1)


def harmonize_nhanes_variables(df: pd.DataFrame) -> pd.DataFrame:
    out = df.copy()

    age_col = get_first_existing_column(out, ["RIDAGEYR"])
    sex_col = get_first_existing_column(out, ["RIAGENDR"])
    race_col = get_first_existing_column(out, ["RIDRETH3", "RIDRETH1"])
    edu_col = get_first_existing_column(out, ["DMDEDUC2", "DMDEDUC3"])
    pir_col = get_first_existing_column(out, ["INDFMPIR"])

    bmi_col = get_first_existing_column(out, ["BMXBMI"])
    waist_col = get_first_existing_column(out, ["BMXWAIST"])
    height_col = get_first_existing_column(out, ["BMXHT"])
    weight_col = get_first_existing_column(out, ["BMXWT"])

    hba1c_col = get_first_existing_column(out, ["LBXGH"])
    fasting_glucose_col = get_first_existing_column(out, ["LBXGLU"])
    ogtt_col = get_first_existing_column(out, ["LBXGLT"])

    known_diabetes_col = get_first_existing_column(out, ["DIQ010"])
    insulin_col = get_first_existing_column(out, ["DIQ050"])
    diabetes_pills_col = get_first_existing_column(out, ["DIQ070"])

    out["age"] = pd.to_numeric(out[age_col], errors="coerce") if age_col else np.nan
    out["sex"] = recode_sex(out[sex_col]) if sex_col else np.nan
    out["race_ethnicity_code"] = pd.to_numeric(out[race_col], errors="coerce") if race_col else np.nan
    out["education_code"] = pd.to_numeric(out[edu_col], errors="coerce") if edu_col else np.nan
    out["income_poverty_ratio"] = pd.to_numeric(out[pir_col], errors="coerce") if pir_col else np.nan

    out["bmi"] = pd.to_numeric(out[bmi_col], errors="coerce") if bmi_col else np.nan
    out["waist_cm"] = pd.to_numeric(out[waist_col], errors="coerce") if waist_col else np.nan
    out["height_cm"] = pd.to_numeric(out[height_col], errors="coerce") if height_col else np.nan
    out["weight_kg"] = pd.to_numeric(out[weight_col], errors="coerce") if weight_col else np.nan

    out["systolic_bp"] = row_mean_existing(out, ["BPXSY1", "BPXSY2", "BPXSY3", "BPXSY4"])
    out["diastolic_bp"] = row_mean_existing(out, ["BPXDI1", "BPXDI2", "BPXDI3", "BPXDI4"])

    out["hba1c"] = pd.to_numeric(out[hba1c_col], errors="coerce") if hba1c_col else np.nan
    out["fasting_glucose"] = pd.to_numeric(out[fasting_glucose_col], errors="coerce") if fasting_glucose_col else np.nan
    out["two_hour_glucose"] = pd.to_numeric(out[ogtt_col], errors="coerce") if ogtt_col else np.nan

    out["known_diabetes"] = recode_yes_no(out[known_diabetes_col]) if known_diabetes_col else np.nan
    out["uses_insulin"] = recode_yes_no(out[insulin_col]) if insulin_col else np.nan
    out["uses_diabetes_pills"] = recode_yes_no(out[diabetes_pills_col]) if diabetes_pills_col else np.nan

    out["age_group"] = make_age_group(out["age"])
    out["bmi_group"] = make_bmi_group(out["bmi"])

    out["overweight_or_obesity"] = np.where(out["bmi"].notna(), (out["bmi"] >= 25).astype(int), np.nan)
    out["obesity"] = np.where(out["bmi"].notna(), (out["bmi"] >= 30).astype(int), np.nan)

    out["hypertension_range_bp"] = np.where(
        out[["systolic_bp", "diastolic_bp"]].notna().any(axis=1),
        ((out["systolic_bp"] >= 130) | (out["diastolic_bp"] >= 80)).astype(int),
        np.nan
    )

    out["uspstf_age_bmi_eligible"] = np.where(
        out[["age", "bmi"]].notna().all(axis=1),
        (((out["age"] >= 35) & (out["age"] <= 70) & (out["bmi"] >= 25))).astype(int),
        np.nan
    )

    return out


nhanes_harmonized = harmonize_nhanes_variables(nhanes_merged_raw)

memory_usage_report(nhanes_harmonized, "nhanes_harmonized")

display(nhanes_harmonized[
    [
        "SEQN",
        "cycle",
        "age",
        "sex",
        "bmi",
        "systolic_bp",
        "diastolic_bp",
        "hba1c",
        "fasting_glucose",
        "two_hour_glucose",
        "known_diabetes"
    ]
].head())

harmonized_path = INTERIM_DATA_DIR / "step6_nhanes_harmonized.parquet"
nhanes_harmonized.to_parquet(harmonized_path, index=False)

print(f"Harmonized panel saved to: {harmonized_path}")
```

```python
def construct_diabetes_outcomes(df: pd.DataFrame) -> pd.DataFrame:
    out = df.copy()

    hba1c_positive = out["hba1c"] >= 6.5
    fasting_glucose_positive = out["fasting_glucose"] >= 126
    ogtt_positive = out["two_hour_glucose"] >= 200
    known_diabetes_positive = out["known_diabetes"] == 1

    hba1c_available = out["hba1c"].notna()
    fasting_glucose_available = out["fasting_glucose"].notna()
    ogtt_available = out["two_hour_glucose"].notna()
    known_diabetes_available = out["known_diabetes"].notna()

    out["has_hba1c"] = hba1c_available.astype(int)
    out["has_fasting_glucose"] = fasting_glucose_available.astype(int)
    out["has_ogtt"] = ogtt_available.astype(int)
    out["has_any_lab"] = (
        hba1c_available
        | fasting_glucose_available
        | ogtt_available
    ).astype(int)

    out["diabetes_binary"] = np.where(
        hba1c_available | fasting_glucose_available | ogtt_available | known_diabetes_available,
        (
            hba1c_positive
            | fasting_glucose_positive
            | ogtt_positive
            | known_diabetes_positive
        ).astype(int),
        np.nan
    )

    out["diabetes_binary_no_ogtt"] = np.where(
        hba1c_available | fasting_glucose_available | known_diabetes_available,
        (
            hba1c_positive
            | fasting_glucose_positive
            | known_diabetes_positive
        ).astype(int),
        np.nan
    )

    dysglycemia_hba1c = out["hba1c"] >= 5.7
    dysglycemia_fasting = out["fasting_glucose"] >= 100
    dysglycemia_ogtt = out["two_hour_glucose"] >= 140

    out["dysglycemia_binary"] = np.where(
        hba1c_available | fasting_glucose_available | ogtt_available,
        (
            dysglycemia_hba1c
            | dysglycemia_fasting
            | dysglycemia_ogtt
        ).astype(int),
        np.nan
    )

    out["undiagnosed_diabetes_binary"] = np.where(
        out["diabetes_binary"].notna() & out["known_diabetes"].notna(),
        ((out["diabetes_binary"] == 1) & (out["known_diabetes"] != 1)).astype(int),
        np.nan
    )

    return out


nhanes_with_outcomes = construct_diabetes_outcomes(nhanes_harmonized)

outcome_cols = [
    "diabetes_binary",
    "diabetes_binary_no_ogtt",
    "dysglycemia_binary",
    "undiagnosed_diabetes_binary",
    "has_hba1c",
    "has_fasting_glucose",
    "has_ogtt",
    "has_any_lab"
]

outcome_summary = []

for col in outcome_cols:
    if col in nhanes_with_outcomes.columns:
        valid = nhanes_with_outcomes[col].dropna()

        outcome_summary.append({
            "variable": col,
            "non_missing": int(valid.shape[0]),
            "mean_or_rate": float(valid.mean()) if len(valid) > 0 else np.nan,
            "positive_count": int((valid == 1).sum()) if len(valid) > 0 else 0
        })

outcome_summary = pd.DataFrame(outcome_summary)

display(outcome_summary)

outcome_summary_path = TABLE_DIR / "step6_outcome_summary.csv"
outcome_summary.to_csv(outcome_summary_path, index=False)

outcome_path = INTERIM_DATA_DIR / "step6_nhanes_harmonized_with_outcomes.parquet"
nhanes_with_outcomes.to_parquet(outcome_path, index=False)

print(f"Outcome summary saved to: {outcome_summary_path}")
print(f"Panel with outcomes saved to: {outcome_path}")
```

```python
def create_initial_screening_cohort(df: pd.DataFrame) -> pd.DataFrame:
    cohort = df.copy()

    cohort = cohort[cohort["age"].notna()].copy()
    cohort = cohort[cohort["age"] >= CONFIG.data.min_age].copy()

    cohort = cohort[cohort["diabetes_binary_no_ogtt"].notna()].copy()

    cohort["primary_target"] = cohort["diabetes_binary_no_ogtt"].astype(int)

    cohort = reduce_dataframe_memory(cohort)

    return cohort


initial_screening_cohort = create_initial_screening_cohort(nhanes_with_outcomes)

memory_usage_report(initial_screening_cohort, "initial_screening_cohort")

cohort_summary = {
    "rows": int(initial_screening_cohort.shape[0]),
    "columns": int(initial_screening_cohort.shape[1]),
    "cycles": sorted(initial_screening_cohort["cycle"].dropna().unique().tolist()),
    "target_positive_count": int(initial_screening_cohort["primary_target"].sum()),
    "target_prevalence": float(initial_screening_cohort["primary_target"].mean()),
    "has_hba1c_rate": float(initial_screening_cohort["has_hba1c"].mean()),
    "has_fasting_glucose_rate": float(initial_screening_cohort["has_fasting_glucose"].mean()),
    "has_ogtt_rate": float(initial_screening_cohort["has_ogtt"].mean()),
    "adult_min_age": CONFIG.data.min_age
}

print(json.dumps(cohort_summary, indent=2))

cohort_path = PROCESSED_DATA_DIR / "initial_screening_cohort.parquet"
initial_screening_cohort.to_parquet(cohort_path, index=False)

save_json(cohort_summary, REPORT_DIR / "step6_initial_screening_cohort_summary.json")

print(f"Initial screening cohort saved to: {cohort_path}")
```

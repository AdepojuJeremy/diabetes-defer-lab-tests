```python
from pathlib import Path
from typing import Dict, List, Tuple, Optional, Any
import json
import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

pd.set_option("io.parquet.engine", "fastparquet")

BASE_DIR = Path.home() / "projects" / "diabetes-defer-lab-tests"

DATA_DIR = BASE_DIR / "data"
INTERIM_DATA_DIR = DATA_DIR / "interim"
PROCESSED_DATA_DIR = DATA_DIR / "processed"

OUTPUT_DIR = BASE_DIR / "outputs"
TABLE_DIR = OUTPUT_DIR / "tables"
REPORT_DIR = BASE_DIR / "reports"

for directory in [
    INTERIM_DATA_DIR,
    PROCESSED_DATA_DIR,
    TABLE_DIR,
    REPORT_DIR
]:
    directory.mkdir(parents=True, exist_ok=True)

def save_json(obj, path):
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)

    with open(path, "w") as f:
        json.dump(obj, f, indent=2, default=str)

    return path

print("Step 7 environment ready.")
```

```python
candidate_paths = [
    PROCESSED_DATA_DIR / "initial_screening_cohort.parquet",
    INTERIM_DATA_DIR / "step6_nhanes_harmonized_with_outcomes.parquet"
]

cohort_path = None

for path in candidate_paths:
    if path.exists():
        cohort_path = path
        break

if cohort_path is None:
    raise FileNotFoundError(
        "Could not find Step 6 output. Check the processed/interim folders."
    )

nhanes = pd.read_parquet(
    cohort_path,
    engine="fastparquet"
)

print("Loaded:", cohort_path)
print("Shape:", nhanes.shape)

display(nhanes.head())
```

```python
def construct_screen_detected_target(df):
    out = df.copy()

    hba1c_available = out["hba1c"].notna()
    fasting_available = out["fasting_glucose"].notna()

    has_primary_lab = hba1c_available | fasting_available

    lab_positive = (
        (out["hba1c"] >= 6.5)
        | (out["fasting_glucose"] >= 126)
    )

    if "known_diabetes" in out.columns:
        not_known_diabetes = (
            out["known_diabetes"].isna()
            | (out["known_diabetes"] != 1)
        )
    else:
        not_known_diabetes = pd.Series(True, index=out.index)

    out["has_primary_diagnostic_lab"] = has_primary_lab.astype(int)

    out["screen_detected_diabetes"] = np.where(
        has_primary_lab & not_known_diabetes,
        lab_positive.astype(int),
        np.nan
    )

    return out


nhanes = construct_screen_detected_target(nhanes)

screening_cohort = nhanes[
    nhanes["screen_detected_diabetes"].notna()
].copy()

screening_cohort["screen_detected_diabetes"] = (
    screening_cohort["screen_detected_diabetes"]
    .astype(int)
)

print("Primary screening cohort:", screening_cohort.shape)
print(
    "Screen-detected diabetes prevalence:",
    round(screening_cohort["screen_detected_diabetes"].mean(), 4)
)
```

```python
CHEAP_FEATURE_GROUPS = {
    "demographics": [
        "age",
        "sex",
        "race_ethnicity_code",
        "education_code",
        "income_poverty_ratio"
    ],

    "anthropometrics": [
        "bmi",
        "waist_cm",
        "height_cm",
        "weight_kg"
    ],

    "blood_pressure": [
        "systolic_bp",
        "diastolic_bp",
        "hypertension_range_bp"
    ],

    "screening_flags": [
        "overweight_or_obesity",
        "obesity",
        "uspstf_age_bmi_eligible"
    ],

    "lifestyle_questionnaire": [
        "SMQ020",
        "SMQ040",
        "PAQ605",
        "PAQ620",
        "PAQ635",
        "PAQ650",
        "ALQ101"
    ],

    "medical_history_non_diabetes": [
        "BPQ020",
        "MCQ160A",
        "MCQ160B",
        "MCQ160C",
        "MCQ160D",
        "MCQ160E",
        "MCQ160F"
    ]
}

EXPENSIVE_LAB_FEATURES = [
    "hba1c",
    "fasting_glucose",
    "two_hour_glucose"
]

LEAKAGE_COLUMNS = [
    "known_diabetes",
    "uses_insulin",
    "uses_diabetes_pills",
    "DIQ010",
    "DIQ050",
    "DIQ070",
    "diabetes_binary",
    "diabetes_binary_no_ogtt",
    "dysglycemia_binary",
    "undiagnosed_diabetes_binary",
    "primary_target",
    "screen_detected_diabetes",
    "has_hba1c",
    "has_fasting_glucose",
    "has_ogtt",
    "has_any_lab",
    "has_primary_diagnostic_lab"
]

def flatten_unique(dictionary):
    output = []

    for values in dictionary.values():
        for value in values:
            if value not in output:
                output.append(value)

    return output


cheap_candidates = flatten_unique(CHEAP_FEATURE_GROUPS)

cheap_features = [
    col for col in cheap_candidates
    if col in screening_cohort.columns
    and col not in LEAKAGE_COLUMNS
]

lab_features = [
    col for col in EXPENSIVE_LAB_FEATURES
    if col in screening_cohort.columns
]

full_features = cheap_features + lab_features

print("Cheap features:", len(cheap_features))
print("Lab features:", len(lab_features))
print("Full features:", len(full_features))

print("\nCheap:")
print(cheap_features)

print("\nLabs:")
print(lab_features)
```

```python
def build_feature_availability_table(df, cheap_features, lab_features):
    records = []

    for feature in cheap_features:
        records.append({
            "feature": feature,
            "tier": "cheap",
            "dtype": str(df[feature].dtype),
            "non_missing": int(df[feature].notna().sum()),
            "missing_rate": float(df[feature].isna().mean()),
            "unique_values": int(df[feature].nunique(dropna=True))
        })

    for feature in lab_features:
        records.append({
            "feature": feature,
            "tier": "expensive_lab",
            "dtype": str(df[feature].dtype),
            "non_missing": int(df[feature].notna().sum()),
            "missing_rate": float(df[feature].isna().mean()),
            "unique_values": int(df[feature].nunique(dropna=True))
        })

    return pd.DataFrame(records)


feature_availability = build_feature_availability_table(
    screening_cohort,
    cheap_features,
    lab_features
)

display(feature_availability)

feature_availability.to_csv(
    TABLE_DIR / "step7_feature_availability.csv",
    index=False
)
```

```python
FORCE_CATEGORICAL = {
    "sex",
    "race_ethnicity_code",
    "education_code",
    "hypertension_range_bp",
    "overweight_or_obesity",
    "obesity",
    "uspstf_age_bmi_eligible",
    "SMQ020",
    "SMQ040",
    "PAQ605",
    "PAQ620",
    "PAQ635",
    "PAQ650",
    "ALQ101",
    "BPQ020",
    "MCQ160A",
    "MCQ160B",
    "MCQ160C",
    "MCQ160D",
    "MCQ160E",
    "MCQ160F"
}

cheap_categorical_features = [
    col for col in cheap_features
    if col in FORCE_CATEGORICAL
    or screening_cohort[col].dtype == "object"
    or str(screening_cohort[col].dtype).startswith("category")
]

cheap_numeric_features = [
    col for col in cheap_features
    if col not in cheap_categorical_features
]

lab_numeric_features = lab_features.copy()

full_numeric_features = cheap_numeric_features + lab_numeric_features
full_categorical_features = cheap_categorical_features.copy()

print("Cheap numeric:", cheap_numeric_features)
print("Cheap categorical:", cheap_categorical_features)
print("Lab numeric:", lab_numeric_features)
```

```python
TARGET_COL = "screen_detected_diabetes"

TRAIN_CYCLES = ["2011-2012", "2013-2014"]
VALIDATION_CYCLES = ["2015-2016"]
TEST_CYCLES = ["2017-2018"]

train_df = screening_cohort[
    screening_cohort["cycle"].isin(TRAIN_CYCLES)
].copy()

val_df = screening_cohort[
    screening_cohort["cycle"].isin(VALIDATION_CYCLES)
].copy()

test_df = screening_cohort[
    screening_cohort["cycle"].isin(TEST_CYCLES)
].copy()

split_summary = pd.DataFrame([
    {
        "split": "train",
        "cycles": ", ".join(TRAIN_CYCLES),
        "n": len(train_df),
        "positives": int(train_df[TARGET_COL].sum()),
        "prevalence": float(train_df[TARGET_COL].mean())
    },
    {
        "split": "validation",
        "cycles": ", ".join(VALIDATION_CYCLES),
        "n": len(val_df),
        "positives": int(val_df[TARGET_COL].sum()),
        "prevalence": float(val_df[TARGET_COL].mean())
    },
    {
        "split": "test",
        "cycles": ", ".join(TEST_CYCLES),
        "n": len(test_df),
        "positives": int(test_df[TARGET_COL].sum()),
        "prevalence": float(test_df[TARGET_COL].mean())
    }
])

display(split_summary)

split_summary.to_csv(
    TABLE_DIR / "step7_temporal_split_summary.csv",
    index=False
)

assert len(train_df) > 0
assert len(val_df) > 0
assert len(test_df) > 0

assert train_df["SEQN"].isin(val_df["SEQN"]).sum() == 0
assert train_df["SEQN"].isin(test_df["SEQN"]).sum() == 0
assert val_df["SEQN"].isin(test_df["SEQN"]).sum() == 0

print("Temporal split checks passed.")
```

```python
X_train_cheap = train_df[cheap_features].copy()
X_val_cheap = val_df[cheap_features].copy()
X_test_cheap = test_df[cheap_features].copy()

X_train_full = train_df[full_features].copy()
X_val_full = val_df[full_features].copy()
X_test_full = test_df[full_features].copy()

y_train = train_df[TARGET_COL].astype(int).copy()
y_val = val_df[TARGET_COL].astype(int).copy()
y_test = test_df[TARGET_COL].astype(int).copy()

print("Cheap train:", X_train_cheap.shape)
print("Full train:", X_train_full.shape)

print("Cheap validation:", X_val_cheap.shape)
print("Full validation:", X_val_full.shape)

print("Cheap test:", X_test_cheap.shape)
print("Full test:", X_test_full.shape)

print("\nClass balance")
print("Train:", y_train.mean())
print("Validation:", y_val.mean())
print("Test:", y_test.mean())
```

```python
def make_one_hot_encoder():
    try:
        return OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        )
    except TypeError:
        return OneHotEncoder(
            handle_unknown="ignore",
            sparse=False
        )


def build_preprocessor(
    numeric_features,
    categorical_features
):
    transformers = []

    if len(numeric_features) > 0:
        numeric_pipeline = Pipeline([
            (
                "imputer",
                SimpleImputer(strategy="median")
            ),
            (
                "scaler",
                StandardScaler()
            )
        ])

        transformers.append(
            (
                "numeric",
                numeric_pipeline,
                numeric_features
            )
        )

    if len(categorical_features) > 0:
        categorical_pipeline = Pipeline([
            (
                "imputer",
                SimpleImputer(strategy="most_frequent")
            ),
            (
                "onehot",
                make_one_hot_encoder()
            )
        ])

        transformers.append(
            (
                "categorical",
                categorical_pipeline,
                categorical_features
            )
        )

    return ColumnTransformer(
        transformers=transformers,
        remainder="drop"
    )


cheap_preprocessor = build_preprocessor(
    cheap_numeric_features,
    cheap_categorical_features
)

full_preprocessor = build_preprocessor(
    full_numeric_features,
    full_categorical_features
)

print("Preprocessors defined.")
print("They will be fitted using TRAIN data only.")
```

```python
modeling_cohort = screening_cohort.copy()

modeling_cohort["split"] = "unused"

modeling_cohort.loc[
    modeling_cohort["cycle"].isin(TRAIN_CYCLES),
    "split"
] = "train"

modeling_cohort.loc[
    modeling_cohort["cycle"].isin(VALIDATION_CYCLES),
    "split"
] = "validation"

modeling_cohort.loc[
    modeling_cohort["cycle"].isin(TEST_CYCLES),
    "split"
] = "test"

modeling_cohort_path = (
    PROCESSED_DATA_DIR
    / "step7_primary_modeling_cohort.parquet"
)

modeling_cohort.to_parquet(
    modeling_cohort_path,
    engine="fastparquet",
    index=False
)

print("Saved:", modeling_cohort_path)
print("Shape:", modeling_cohort.shape)
```

```python
feature_metadata = {
    "target": TARGET_COL,

    "primary_target_definition": {
        "positive": (
            "HbA1c >= 6.5% OR fasting plasma glucose >= 126 mg/dL"
        ),
        "population": (
            "Adults without known diabetes and with at least one "
            "primary diagnostic laboratory measurement"
        ),
        "ogtt_primary_target": False
    },

    "cheap_features": cheap_features,
    "cheap_numeric_features": cheap_numeric_features,
    "cheap_categorical_features": cheap_categorical_features,

    "lab_features": lab_features,

    "full_features": full_features,
    "full_numeric_features": full_numeric_features,
    "full_categorical_features": full_categorical_features,

    "leakage_excluded_columns": LEAKAGE_COLUMNS,

    "split_strategy": "temporal NHANES cycle split",

    "train_cycles": TRAIN_CYCLES,
    "validation_cycles": VALIDATION_CYCLES,
    "test_cycles": TEST_CYCLES,

    "train_n": int(len(train_df)),
    "validation_n": int(len(val_df)),
    "test_n": int(len(test_df))
}

metadata_path = REPORT_DIR / "step7_feature_metadata.json"

save_json(
    feature_metadata,
    metadata_path
)

print(json.dumps(feature_metadata, indent=2))
```

```python
cheap_leakage = sorted(
    set(cheap_features)
    .intersection(set(LEAKAGE_COLUMNS))
)

lab_in_cheap = sorted(
    set(cheap_features)
    .intersection(set(lab_features))
)

print("Leakage columns found in cheap features:", cheap_leakage)
print("Lab variables found in cheap features:", lab_in_cheap)

assert len(cheap_leakage) == 0
assert len(lab_in_cheap) == 0

print("Cheap-stage leakage audit passed.")
```

```python
step7_checks = {
    "modeling_cohort_rows": int(len(modeling_cohort)),
    "train_rows": int(len(train_df)),
    "validation_rows": int(len(val_df)),
    "test_rows": int(len(test_df)),

    "train_prevalence": float(y_train.mean()),
    "validation_prevalence": float(y_val.mean()),
    "test_prevalence": float(y_test.mean()),

    "cheap_feature_count": int(len(cheap_features)),
    "lab_feature_count": int(len(lab_features)),
    "full_feature_count": int(len(full_features)),

    "cheap_leakage_columns": cheap_leakage,
    "lab_variables_in_cheap_tier": lab_in_cheap,

    "modeling_cohort_saved": modeling_cohort_path.exists(),
    "metadata_saved": metadata_path.exists()
}

save_json(
    step7_checks,
    REPORT_DIR / "step7_completion_checks.json"
)

print(json.dumps(step7_checks, indent=2))

if (
    step7_checks["train_rows"] > 0
    and step7_checks["validation_rows"] > 0
    and step7_checks["test_rows"] > 0
    and step7_checks["cheap_feature_count"] > 0
    and len(cheap_leakage) == 0
    and len(lab_in_cheap) == 0
):
    print("Step 7 completed successfully.")
else:
    print("Step 7 completed with warnings.")
```

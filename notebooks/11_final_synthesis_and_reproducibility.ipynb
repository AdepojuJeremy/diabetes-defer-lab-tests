````python
def save_json(
    obj,
    path
):
    path = Path(path)

    path.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    with open(
        path,
        "w",
        encoding="utf-8"
    ) as f:
        json.dump(
            obj,
            f,
            indent=2,
            default=str
        )

    return path


def load_json(
    path,
    default=None
):
    path = Path(path)

    if not path.exists():
        return default

    with open(
        path,
        "r",
        encoding="utf-8"
    ) as f:
        return json.load(f)


def load_csv_if_exists(
    path
):
    path = Path(path)

    if not path.exists():
        return None

    return pd.read_csv(
        path
    )


def copy_if_exists(
    source,
    destination
):
    source = Path(source)
    destination = Path(destination)

    if not source.exists():
        return False

    destination.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    shutil.copy2(
        source,
        destination
    )

    return True


def file_sha256(
    path,
    chunk_size=1024 * 1024
):
    path = Path(path)

    digest = hashlib.sha256()

    with open(
        path,
        "rb"
    ) as f:

        while True:

            chunk = f.read(
                chunk_size
            )

            if not chunk:
                break

            digest.update(
                chunk
            )

    return digest.hexdigest()


def save_plotly_final(
    fig,
    name,
    width=1300,
    height=800
):
    html_path = (
        FINAL_FIGURE_DIR
        / f"{name}.html"
    )

    png_path = (
        FINAL_FIGURE_DIR
        / f"{name}.png"
    )

    fig.write_html(
        html_path
    )

    try:
        fig.write_image(
            png_path,
            width=width,
            height=height,
            scale=2
        )

        print(
            "PNG saved:",
            png_path
        )

    except Exception as error:

        print(
            f"PNG export skipped for "
            f"{name}: {error}"
        )

        png_path = None

    return {
        "html": str(
            html_path
        ),

        "png": (
            str(
                png_path
            )
            if png_path is not None
            else None
        )
    }


def safe_divide(
    numerator,
    denominator
):
    if denominator == 0:
        return np.nan

    return numerator / denominator
````

````python
step12_dataset_path = (
    PROCESSED_DATA_DIR
    / "step12_fairness_analysis_dataset.parquet"
)

step11_assignment_path = (
    PROCESSED_DATA_DIR
    / "step11_budget_assignments.parquet"
)

cohort_path = (
    PROCESSED_DATA_DIR
    / "step7_primary_modeling_cohort.parquet"
)

if step12_dataset_path.exists():

    final_test_df = pd.read_parquet(
        step12_dataset_path,
        engine="fastparquet"
    )

    final_test_source = (
        "Step 12 fairness analysis dataset"
    )

elif step11_assignment_path.exists():

    final_test_df = pd.read_parquet(
        step11_assignment_path,
        engine="fastparquet"
    )

    final_test_source = (
        "Step 11 budget assignments"
    )

else:

    raise FileNotFoundError(
        "Neither the Step 12 analysis dataset "
        "nor Step 11 assignment dataset exists."
    )

print(
    "Final test source:",
    final_test_source
)

print(
    "Rows:",
    len(
        final_test_df
    )
)

print(
    "Columns:",
    len(
        final_test_df.columns
    )
)
````

````python
required_final_columns = [
    TARGET_COL,
    "cascade_prediction",
    "referred"
]

missing_final_columns = [
    column
    for column in required_final_columns
    if column not in final_test_df.columns
]

if missing_final_columns:

    raise RuntimeError(
        "Final test dataset is missing: "
        + ", ".join(
            missing_final_columns
        )
    )


y_true = (
    final_test_df[
        TARGET_COL
    ]
    .astype(int)
    .to_numpy()
)

y_pred = (
    final_test_df[
        "cascade_prediction"
    ]
    .astype(int)
    .to_numpy()
)

referred = (
    final_test_df[
        "referred"
    ]
    .astype(int)
    .to_numpy()
)

tn, fp, fn, tp = (
    confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1]
    )
    .ravel()
)

FINAL_PRIMARY_METRICS = {
    "n_test": int(
        len(
            y_true
        )
    ),

    "positive_cases": int(
        y_true.sum()
    ),

    "prevalence": float(
        y_true.mean()
    ),

    "referral_rate": float(
        referred.mean()
    ),

    "accuracy": float(
        accuracy_score(
            y_true,
            y_pred
        )
    ),

    "balanced_accuracy": float(
        balanced_accuracy_score(
            y_true,
            y_pred
        )
    ),

    "sensitivity": float(
        recall_score(
            y_true,
            y_pred
        )
    ),

    "specificity": float(
        safe_divide(
            tn,
            tn + fp
        )
    ),

    "precision": float(
        precision_score(
            y_true,
            y_pred,
            zero_division=0
        )
    ),

    "f1": float(
        f1_score(
            y_true,
            y_pred,
            zero_division=0
        )
    ),

    "true_positive": int(tp),
    "true_negative": int(tn),
    "false_positive": int(fp),
    "false_negative": int(fn),

    "missed_case_rate": float(
        safe_divide(
            fn,
            tp + fn
        )
    )
}

save_json(
    FINAL_PRIMARY_METRICS,
    FINAL_METADATA_DIR
    / "final_primary_35pct_metrics.json"
)

print(
    json.dumps(
        FINAL_PRIMARY_METRICS,
        indent=2
    )
)
````

````python
BOOTSTRAP_REPLICATES = 1000
BOOTSTRAP_SEED = 2025

rng = np.random.default_rng(
    BOOTSTRAP_SEED
)

bootstrap_records = []

n_test = len(
    y_true
)

for replicate in range(
    BOOTSTRAP_REPLICATES
):

    indices = rng.integers(
        0,
        n_test,
        size=n_test
    )

    y_boot = (
        y_true[
            indices
        ]
    )

    pred_boot = (
        y_pred[
            indices
        ]
    )

    referral_boot = (
        referred[
            indices
        ]
    )

    if len(
        np.unique(
            y_boot
        )
    ) < 2:
        continue

    tn_b, fp_b, fn_b, tp_b = (
        confusion_matrix(
            y_boot,
            pred_boot,
            labels=[0, 1]
        )
        .ravel()
    )

    bootstrap_records.append({
        "referral_rate": float(
            referral_boot.mean()
        ),

        "sensitivity": float(
            safe_divide(
                tp_b,
                tp_b + fn_b
            )
        ),

        "specificity": float(
            safe_divide(
                tn_b,
                tn_b + fp_b
            )
        ),

        "balanced_accuracy": float(
            balanced_accuracy_score(
                y_boot,
                pred_boot
            )
        ),

        "precision": float(
            precision_score(
                y_boot,
                pred_boot,
                zero_division=0
            )
        )
    })


final_bootstrap = pd.DataFrame(
    bootstrap_records
)

ci_records = []

for metric in [
    "referral_rate",
    "sensitivity",
    "specificity",
    "balanced_accuracy",
    "precision"
]:

    ci_records.append({
        "metric": metric,

        "estimate": (
            FINAL_PRIMARY_METRICS[
                metric
            ]
        ),

        "ci_2.5": float(
            final_bootstrap[
                metric
            ].quantile(
                0.025
            )
        ),

        "ci_97.5": float(
            final_bootstrap[
                metric
            ].quantile(
                0.975
            )
        )
    })


final_confidence_intervals = (
    pd.DataFrame(
        ci_records
    )
)

display(
    final_confidence_intervals
)

final_confidence_intervals.to_csv(
    FINAL_TABLE_DIR
    / "table_final_primary_performance_with_95ci.csv",
    index=False
)
````

````python
limitations_text = """
# Limitations and Interpretation

1. NHANES is a cross-sectional survey. The analysis therefore evaluates screening and diagnostic classification rather than longitudinal progression to diabetes.

2. The primary endpoint is defined using laboratory measurements that are also supplied to the full-information branch. Interpretation of the full laboratory predictor is consequently not emphasized because HbA1c and fasting glucose directly define the endpoint.

3. Laboratory referral cost is represented using normalized relative costs rather than a single universal monetary value. The cost-sensitivity analysis is therefore more informative than any single assumed cost ratio.

4. The learned gate optimizes expected value under a constrained laboratory budget. Equal referral rates across demographic groups are not imposed as an objective because clinical need and disease prevalence can differ between groups. Equity conclusions should therefore consider positive-case referral coverage, under-referral, sensitivity, specificity, and calibration jointly.

5. The temporal split provides a stronger test than random splitting but does not constitute geographic or health-system external validation.

6. NHANES questionnaire and lifestyle measures may contain recall error, measurement error, or missingness.

7. The model should not be interpreted as a replacement for clinical diagnosis or existing screening guidelines. It is a resource-allocation framework for investigating when inexpensive screening information may justify additional laboratory testing.

8. Deployment would require prospective validation, externally defined laboratory costs, operational workflow analysis, and assessment under the target population's disease prevalence and clinical practice patterns.

9. The oracle referral strategy is strictly an analytical upper bound and is not deployable because it relies on information unavailable at the actual referral decision point.

10. Integrated Gradients and ablation analyses characterize model dependence and attribution but do not establish causal effects of patient characteristics on diabetes or referral decisions.
""".strip()

limitations_path = (
    FINAL_DIR
    / "FINAL_LIMITATIONS.md"
)

with open(
    limitations_path,
    "w",
    encoding="utf-8"
) as f:

    f.write(
        limitations_text
    )

print(
    limitations_text
)
````

````python
inventory_roots = [
    NOTEBOOK_DIR,
    TABLE_DIR,
    FIGURE_DIR,
    MODEL_DIR,
    REPORT_DIR
]

artifact_records = []

for root in inventory_roots:

    if not root.exists():
        continue

    for path in root.rglob(
        "*"
    ):

        if not path.is_file():
            continue

        try:

            relative_path = (
                path.relative_to(
                    BASE_DIR
                )
            )

        except ValueError:

            relative_path = path

        artifact_records.append({
            "relative_path": str(
                relative_path
            ),

            "suffix": (
                path.suffix.lower()
            ),

            "size_bytes": int(
                path.stat().st_size
            ),

            "modified_utc": (
                datetime.fromtimestamp(
                    path.stat().st_mtime,
                    tz=timezone.utc
                ).isoformat()
            )
        })


artifact_inventory = pd.DataFrame(
    artifact_records
)

artifact_inventory = (
    artifact_inventory
    .sort_values(
        "relative_path"
    )
)

display(
    artifact_inventory.head(30)
)

artifact_inventory.to_csv(
    FINAL_METADATA_DIR
    / "complete_artifact_inventory.csv",
    index=False
)

print(
    "Total artifacts:",
    len(
        artifact_inventory
    )
)
````

````python
CHECKSUM_SUFFIXES = {
    ".ipynb",
    ".py",
    ".json",
    ".csv",
    ".md",
    ".txt",
    ".pt",
    ".joblib"
}

checksum_records = []

for _, row in (
    artifact_inventory.iterrows()
):

    relative_path = Path(
        row[
            "relative_path"
        ]
    )

    if (
        relative_path.suffix.lower()
        not in CHECKSUM_SUFFIXES
    ):
        continue

    absolute_path = (
        BASE_DIR
        / relative_path
    )

    if not absolute_path.exists():
        continue

    try:

        sha256 = file_sha256(
            absolute_path
        )

    except Exception as error:

        sha256 = (
            f"ERROR: {error}"
        )

    checksum_records.append({
        "relative_path": str(
            relative_path
        ),

        "sha256": sha256,

        "size_bytes": int(
            absolute_path.stat().st_size
        )
    })


checksum_table = pd.DataFrame(
    checksum_records
)

checksum_table.to_csv(
    FINAL_METADATA_DIR
    / "artifact_sha256_checksums.csv",
    index=False
)

print(
    "Checksummed artifacts:",
    len(
        checksum_table
    )
)
````

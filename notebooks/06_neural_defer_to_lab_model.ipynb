```python
def make_one_hot_encoder():
    try:
        return OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        )
    except TypeError:
        return OneHotEncoder(
            handle_unknown="ignore",
            sparse=False
        )


def build_neural_preprocessor(
    numeric_features,
    categorical_features
):
    transformers = []

    if numeric_features:

        numeric_pipeline = Pipeline([
            (
                "imputer",
                SimpleImputer(
                    strategy="median",
                    add_indicator=True
                )
            ),
            (
                "scaler",
                StandardScaler()
            )
        ])

        transformers.append(
            (
                "numeric",
                numeric_pipeline,
                numeric_features
            )
        )

    if categorical_features:

        categorical_pipeline = Pipeline([
            (
                "imputer",
                SimpleImputer(
                    strategy="most_frequent"
                )
            ),
            (
                "onehot",
                make_one_hot_encoder()
            )
        ])

        transformers.append(
            (
                "categorical",
                categorical_pipeline,
                categorical_features
            )
        )

    return ColumnTransformer(
        transformers=transformers,
        remainder="drop"
    )


cheap_preprocessor = (
    build_neural_preprocessor(
        cheap_numeric_features,
        cheap_categorical_features
    )
)

full_preprocessor = (
    build_neural_preprocessor(
        full_numeric_features,
        full_categorical_features
    )
)

print(
    "Preprocessors created."
)
```

```python
class DeferLabDataset(Dataset):

    def __init__(
        self,
        X_cheap,
        X_full,
        y
    ):
        self.X_cheap = torch.tensor(
            X_cheap,
            dtype=torch.float32
        )

        self.X_full = torch.tensor(
            X_full,
            dtype=torch.float32
        )

        self.y = torch.tensor(
            y,
            dtype=torch.float32
        )

    def __len__(self):
        return len(self.y)

    def __getitem__(
        self,
        index
    ):
        return (
            self.X_cheap[index],
            self.X_full[index],
            self.y[index]
        )
```

```python
BATCH_SIZE = 512

train_dataset = DeferLabDataset(
    X_train_cheap,
    X_train_full,
    y_train
)

val_dataset = DeferLabDataset(
    X_val_cheap,
    X_val_full,
    y_val
)

test_dataset = DeferLabDataset(
    X_test_cheap,
    X_test_full,
    y_test
)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=0,
    pin_memory=USE_AMP
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
    pin_memory=USE_AMP
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
    pin_memory=USE_AMP
)

print(
    "Train batches:",
    len(train_loader)
)

print(
    "Validation batches:",
    len(val_loader)
)

print(
    "Test batches:",
    len(test_loader)
)
```

```python
class MLPBlock(nn.Module):

    def __init__(
        self,
        input_dim,
        hidden_dim=128,
        num_layers=3,
        dropout=0.15
    ):
        super().__init__()

        layers = []

        current_dim = input_dim

        for _ in range(num_layers):

            layers.extend([
                nn.Linear(
                    current_dim,
                    hidden_dim
                ),

                nn.LayerNorm(
                    hidden_dim
                ),

                nn.GELU(),

                nn.Dropout(
                    dropout
                )
            ])

            current_dim = hidden_dim

        self.network = nn.Sequential(
            *layers
        )

    def forward(
        self,
        x
    ):
        return self.network(x)
```

```python
class DeferToLabNet(nn.Module):

    def __init__(
        self,
        cheap_input_dim,
        full_input_dim,
        hidden_dim=128,
        num_layers=3,
        dropout=0.15
    ):
        super().__init__()

        self.cheap_encoder = MLPBlock(
            input_dim=cheap_input_dim,
            hidden_dim=hidden_dim,
            num_layers=num_layers,
            dropout=dropout
        )

        self.full_encoder = MLPBlock(
            input_dim=full_input_dim,
            hidden_dim=hidden_dim,
            num_layers=num_layers,
            dropout=dropout
        )

        self.gate_encoder = MLPBlock(
            input_dim=cheap_input_dim,
            hidden_dim=hidden_dim,
            num_layers=2,
            dropout=dropout
        )

        self.cheap_head = nn.Linear(
            hidden_dim,
            1
        )

        self.full_head = nn.Linear(
            hidden_dim,
            1
        )

        self.defer_head = nn.Linear(
            hidden_dim,
            1
        )

    def forward(
        self,
        x_cheap,
        x_full
    ):

        cheap_representation = (
            self.cheap_encoder(
                x_cheap
            )
        )

        full_representation = (
            self.full_encoder(
                x_full
            )
        )

        gate_representation = (
            self.gate_encoder(
                x_cheap
            )
        )

        cheap_logit = (
            self.cheap_head(
                cheap_representation
            ).squeeze(-1)
        )

        full_logit = (
            self.full_head(
                full_representation
            ).squeeze(-1)
        )

        defer_logit = (
            self.defer_head(
                gate_representation
            ).squeeze(-1)
        )

        defer_probability = (
            torch.sigmoid(
                defer_logit
            )
        )

        return {
            "cheap_logit": cheap_logit,
            "full_logit": full_logit,
            "defer_logit": defer_logit,
            "defer_probability": (
                defer_probability
            )
        }
```

```python
LAB_COST = 1.0

TARGET_BUDGET = 0.35

DEFER_COST_LAMBDA = 0.10

BUDGET_PENALTY_LAMBDA = 10.0

ENTROPY_BONUS_LAMBDA = 0.001


positive_count = (
    y_train.sum()
)

negative_count = (
    len(y_train)
    - positive_count
)

POS_WEIGHT = (
    negative_count
    / max(
        positive_count,
        1
    )
)

POS_WEIGHT_TENSOR = torch.tensor(
    POS_WEIGHT,
    dtype=torch.float32,
    device=DEVICE
)

print(
    "Positive class weight:",
    round(
        float(POS_WEIGHT),
        3
    )
)

print(
    "Target referral budget:",
    TARGET_BUDGET
)
```

```python
def defer_to_lab_loss(
    outputs,
    targets
):

    cheap_logit = outputs[
        "cheap_logit"
    ]

    full_logit = outputs[
        "full_logit"
    ]

    defer_probability = outputs[
        "defer_probability"
    ]

    cheap_loss = (
        F.binary_cross_entropy_with_logits(
            cheap_logit,
            targets,
            pos_weight=POS_WEIGHT_TENSOR,
            reduction="none"
        )
    )

    full_loss = (
        F.binary_cross_entropy_with_logits(
            full_logit,
            targets,
            pos_weight=POS_WEIGHT_TENSOR,
            reduction="none"
        )
    )

    selective_loss = (
        (
            1.0
            - defer_probability
        )
        * cheap_loss
        +
        defer_probability
        * full_loss
    ).mean()

    expected_referral_rate = (
        defer_probability.mean()
    )

    expected_lab_cost = (
        LAB_COST
        * expected_referral_rate
    )

    cost_penalty = (
        DEFER_COST_LAMBDA
        * expected_lab_cost
    )

    budget_violation = (
        torch.relu(
            expected_lab_cost
            - TARGET_BUDGET
        )
    )

    budget_penalty = (
        BUDGET_PENALTY_LAMBDA
        * budget_violation.pow(2)
    )

    eps = 1e-7

    gate_entropy = -(
        defer_probability
        * torch.log(
            defer_probability + eps
        )
        +
        (
            1.0
            - defer_probability
        )
        * torch.log(
            1.0
            - defer_probability
            + eps
        )
    ).mean()

    entropy_bonus = (
        -ENTROPY_BONUS_LAMBDA
        * gate_entropy
    )

    total_loss = (
        selective_loss
        + cost_penalty
        + budget_penalty
        + entropy_bonus
    )

    return {
        "loss": total_loss,
        "selective_loss": (
            selective_loss
        ),
        "cost_penalty": (
            cost_penalty
        ),
        "budget_penalty": (
            budget_penalty
        ),
        "gate_entropy": (
            gate_entropy
        ),
        "expected_referral_rate": (
            expected_referral_rate
        ),
        "cheap_loss": (
            cheap_loss.mean()
        ),
        "full_loss": (
            full_loss.mean()
        )
    }
```

```python
LEARNING_RATE = 1e-3
WEIGHT_DECAY = 1e-4

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)

scheduler = (
    torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode="min",
        factor=0.5,
        patience=4,
        min_lr=1e-6
    )
)

scaler = (
    torch.cuda.amp.GradScaler(
        enabled=USE_AMP
    )
)

print(
    "Optimizer ready."
)
```

```python
def train_one_epoch(
    model,
    loader,
    optimizer,
    scaler
):

    model.train()

    totals = {
        "loss": 0.0,
        "selective_loss": 0.0,
        "cost_penalty": 0.0,
        "budget_penalty": 0.0,
        "gate_entropy": 0.0,
        "expected_referral_rate": 0.0
    }

    sample_count = 0

    for (
        x_cheap,
        x_full,
        targets
    ) in loader:

        x_cheap = x_cheap.to(
            DEVICE,
            non_blocking=True
        )

        x_full = x_full.to(
            DEVICE,
            non_blocking=True
        )

        targets = targets.to(
            DEVICE,
            non_blocking=True
        )

        batch_size = (
            targets.shape[0]
        )

        optimizer.zero_grad(
            set_to_none=True
        )

        with torch.cuda.amp.autocast(
            enabled=USE_AMP
        ):

            outputs = model(
                x_cheap,
                x_full
            )

            losses = (
                defer_to_lab_loss(
                    outputs,
                    targets
                )
            )

        scaler.scale(
            losses["loss"]
        ).backward()

        scaler.unscale_(
            optimizer
        )

        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=5.0
        )

        scaler.step(
            optimizer
        )

        scaler.update()

        sample_count += (
            batch_size
        )

        for key in totals:

            totals[key] += (
                float(
                    losses[key]
                    .detach()
                    .cpu()
                )
                * batch_size
            )

    return {
        key: (
            value
            / sample_count
        )
        for key, value
        in totals.items()
    }
```

```python
@torch.no_grad()
def predict_defer_model(
    model,
    loader
):

    model.eval()

    y_true = []

    cheap_probabilities = []
    full_probabilities = []
    defer_probabilities = []

    total_loss = 0.0
    total_n = 0

    for (
        x_cheap,
        x_full,
        targets
    ) in loader:

        x_cheap = x_cheap.to(
            DEVICE,
            non_blocking=True
        )

        x_full = x_full.to(
            DEVICE,
            non_blocking=True
        )

        targets = targets.to(
            DEVICE,
            non_blocking=True
        )

        with torch.cuda.amp.autocast(
            enabled=USE_AMP
        ):

            outputs = model(
                x_cheap,
                x_full
            )

            losses = (
                defer_to_lab_loss(
                    outputs,
                    targets
                )
            )

        cheap_prob = torch.sigmoid(
            outputs["cheap_logit"]
        )

        full_prob = torch.sigmoid(
            outputs["full_logit"]
        )

        defer_prob = outputs[
            "defer_probability"
        ]

        batch_n = (
            len(targets)
        )

        total_loss += (
            float(
                losses["loss"]
                .detach()
                .cpu()
            )
            * batch_n
        )

        total_n += batch_n

        y_true.append(
            targets.cpu().numpy()
        )

        cheap_probabilities.append(
            cheap_prob.cpu().numpy()
        )

        full_probabilities.append(
            full_prob.cpu().numpy()
        )

        defer_probabilities.append(
            defer_prob.cpu().numpy()
        )

    return {
        "y_true": np.concatenate(
            y_true
        ),

        "cheap_probability": (
            np.concatenate(
                cheap_probabilities
            )
        ),

        "full_probability": (
            np.concatenate(
                full_probabilities
            )
        ),

        "defer_probability": (
            np.concatenate(
                defer_probabilities
            )
        ),

        "loss": (
            total_loss
            / total_n
        )
    }
```

```python
MAX_EPOCHS = 60

EARLY_STOPPING_PATIENCE = 10

MIN_DELTA = 1e-4

best_validation_loss = np.inf

best_epoch = 0

patience_counter = 0

training_history = []

best_model_path = (
    STEP10_MODEL_DIR
    / "best_neural_defer_model.pt"
)

training_start = time.time()


for epoch in range(
    1,
    MAX_EPOCHS + 1
):

    train_metrics = train_one_epoch(
        model,
        train_loader,
        optimizer,
        scaler
    )

    validation_outputs = (
        predict_defer_model(
            model,
            val_loader
        )
    )

    validation_loss = (
        validation_outputs["loss"]
    )

    scheduler.step(
        validation_loss
    )

    current_lr = (
        optimizer
        .param_groups[0]["lr"]
    )

    record = {
        "epoch": epoch,
        "train_loss": (
            train_metrics["loss"]
        ),
        "validation_loss": (
            validation_loss
        ),
        "selective_loss": (
            train_metrics[
                "selective_loss"
            ]
        ),
        "referral_rate": (
            train_metrics[
                "expected_referral_rate"
            ]
        ),
        "cost_penalty": (
            train_metrics[
                "cost_penalty"
            ]
        ),
        "budget_penalty": (
            train_metrics[
                "budget_penalty"
            ]
        ),
        "gate_entropy": (
            train_metrics[
                "gate_entropy"
            ]
        ),
        "learning_rate": (
            current_lr
        )
    }

    training_history.append(
        record
    )

    if (
        validation_loss
        <
        best_validation_loss
        - MIN_DELTA
    ):

        best_validation_loss = (
            validation_loss
        )

        best_epoch = epoch

        patience_counter = 0

        torch.save(
            {
                "model_state_dict": (
                    model.state_dict()
                ),

                "cheap_input_dim": (
                    CHEAP_INPUT_DIM
                ),

                "full_input_dim": (
                    FULL_INPUT_DIM
                ),

                "hidden_dim": 128,
                "num_layers": 3,
                "dropout": 0.15,

                "target_budget": (
                    TARGET_BUDGET
                ),

                "lab_cost": (
                    LAB_COST
                ),

                "seed": (
                    GLOBAL_SEED
                )
            },
            best_model_path
        )

    else:

        patience_counter += 1

    if (
        epoch == 1
        or epoch % 5 == 0
    ):

        print(
            f"Epoch {epoch:03d} | "
            f"Train {train_metrics['loss']:.4f} | "
            f"Val {validation_loss:.4f} | "
            f"Referral {train_metrics['expected_referral_rate']:.3f} | "
            f"LR {current_lr:.6f}"
        )

    if (
        patience_counter
        >= EARLY_STOPPING_PATIENCE
    ):

        print(
            f"Early stopping at epoch {epoch}."
        )

        break


training_minutes = (
    time.time()
    - training_start
) / 60

print()

print(
    "Best epoch:",
    best_epoch
)

print(
    "Best validation loss:",
    round(
        best_validation_loss,
        5
    )
)

print(
    "Training time:",
    round(
        training_minutes,
        2
    ),
    "minutes"
)
```

```python
def select_defer_threshold_for_budget(
    defer_probability,
    target_budget
):

    probabilities = np.asarray(
        defer_probability
    )

    target_budget = float(
        np.clip(
            target_budget,
            0,
            1
        )
    )

    if target_budget <= 0:
        return float(
            probabilities.max()
            + 1e-6
        )

    if target_budget >= 1:
        return float(
            probabilities.min()
            - 1e-6
        )

    threshold = np.quantile(
        probabilities,
        1.0 - target_budget
    )

    return float(
        threshold
    )


DEFER_THRESHOLD = (
    select_defer_threshold_for_budget(
        val_outputs[
            "defer_probability"
        ],
        TARGET_BUDGET
    )
)

val_referral_mask = (
    val_outputs[
        "defer_probability"
    ]
    >= DEFER_THRESHOLD
)

test_referral_mask = (
    test_outputs[
        "defer_probability"
    ]
    >= DEFER_THRESHOLD
)

print(
    "Defer threshold:",
    round(
        DEFER_THRESHOLD,
        5
    )
)

print(
    "Validation referral rate:",
    round(
        val_referral_mask.mean(),
        4
    )
)

print(
    "Test referral rate:",
    round(
        test_referral_mask.mean(),
        4
    )
)
```

```python
def combine_cascade_probability(
    cheap_probability,
    full_probability,
    referral_mask
):

    cheap_probability = np.asarray(
        cheap_probability
    )

    full_probability = np.asarray(
        full_probability
    )

    referral_mask = np.asarray(
        referral_mask
    ).astype(bool)

    return np.where(
        referral_mask,
        full_probability,
        cheap_probability
    )


val_cascade_probability = (
    combine_cascade_probability(
        val_outputs[
            "cheap_probability"
        ],
        val_outputs[
            "full_probability"
        ],
        val_referral_mask
    )
)

test_cascade_probability = (
    combine_cascade_probability(
        test_outputs[
            "cheap_probability"
        ],
        test_outputs[
            "full_probability"
        ],
        test_referral_mask
    )
)
```

```python
def find_youden_threshold(
    y_true,
    probability
):

    fpr, tpr, thresholds = roc_curve(
        y_true,
        probability
    )

    score = (
        tpr - fpr
    )

    index = np.nanargmax(
        score
    )

    return float(
        thresholds[index]
    )


CLASSIFICATION_THRESHOLD = (
    find_youden_threshold(
        y_val,
        val_cascade_probability
    )
)

print(
    "Classification threshold:",
    round(
        CLASSIFICATION_THRESHOLD,
        5
    )
)
```

```python
def evaluate_prediction(
    y_true,
    probability,
    threshold,
    referral_mask=None
):

    y_true = np.asarray(
        y_true
    ).astype(int)

    probability = np.asarray(
        probability
    )

    prediction = (
        probability
        >= threshold
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        prediction,
        labels=[0, 1]
    ).ravel()

    sensitivity = (
        tp / (tp + fn)
        if (tp + fn) > 0
        else np.nan
    )

    specificity = (
        tn / (tn + fp)
        if (tn + fp) > 0
        else np.nan
    )

    metrics = {
        "auroc": (
            roc_auc_score(
                y_true,
                probability
            )
        ),

        "auprc": (
            average_precision_score(
                y_true,
                probability
            )
        ),

        "accuracy": (
            accuracy_score(
                y_true,
                prediction
            )
        ),

        "balanced_accuracy": (
            balanced_accuracy_score(
                y_true,
                prediction
            )
        ),

        "sensitivity": (
            sensitivity
        ),

        "specificity": (
            specificity
        ),

        "f1": (
            f1_score(
                y_true,
                prediction,
                zero_division=0
            )
        ),

        "brier": (
            brier_score_loss(
                y_true,
                probability
            )
        ),

        "threshold": (
            float(threshold)
        ),

        "tp": int(tp),
        "tn": int(tn),
        "fp": int(fp),
        "fn": int(fn)
    }

    if referral_mask is not None:

        referral_rate = float(
            np.mean(
                referral_mask
            )
        )

        metrics[
            "referral_rate"
        ] = referral_rate

        metrics[
            "average_lab_cost"
        ] = (
            LAB_COST
            * referral_rate
        )

    return metrics
```

```python
cheap_validation_threshold = (
    find_youden_threshold(
        y_val,
        val_outputs[
            "cheap_probability"
        ]
    )
)

full_validation_threshold = (
    find_youden_threshold(
        y_val,
        val_outputs[
            "full_probability"
        ]
    )
)


evaluation_records = []


cheap_metrics = evaluate_prediction(
    y_test,
    test_outputs[
        "cheap_probability"
    ],
    cheap_validation_threshold,
    referral_mask=np.zeros(
        len(y_test),
        dtype=bool
    )
)

evaluation_records.append({
    "model": (
        "neural_cheap_only"
    ),
    **cheap_metrics
})


full_metrics = evaluate_prediction(
    y_test,
    test_outputs[
        "full_probability"
    ],
    full_validation_threshold,
    referral_mask=np.ones(
        len(y_test),
        dtype=bool
    )
)

evaluation_records.append({
    "model": (
        "neural_full_lab"
    ),
    **full_metrics
})


cascade_metrics = evaluate_prediction(
    y_test,
    test_cascade_probability,
    CLASSIFICATION_THRESHOLD,
    referral_mask=(
        test_referral_mask
    )
)

evaluation_records.append({
    "model": (
        "neural_defer_to_lab"
    ),
    **cascade_metrics
})


neural_metrics = pd.DataFrame(
    evaluation_records
)

display(
    neural_metrics
)

neural_metrics.to_csv(
    TABLE_DIR
    / "step10_neural_defer_model_metrics.csv",
    index=False
)
```

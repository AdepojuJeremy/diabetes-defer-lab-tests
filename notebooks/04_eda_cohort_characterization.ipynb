```python
def save_json(obj, path):
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)

    with open(path, "w") as f:
        json.dump(obj, f, indent=2, default=str)

    return path


def save_plotly(fig, name, width=1200, height=750):
    html_path = FIGURE_DIR / f"{name}.html"
    png_path = FIGURE_DIR / f"{name}.png"

    fig.write_html(html_path)

    try:
        fig.write_image(
            png_path,
            width=width,
            height=height,
            scale=2
        )
    except Exception as error:
        print(f"PNG export skipped for {name}: {error}")
        png_path = None

    return {
        "html": html_path,
        "png": png_path
    }


def safe_existing(df, columns):
    return [col for col in columns if col in df.columns]


def safe_rate(series):
    series = pd.to_numeric(series, errors="coerce").dropna()

    if len(series) == 0:
        return np.nan

    return float(series.mean())


def median_iqr(series):
    x = pd.to_numeric(series, errors="coerce").dropna()

    if len(x) == 0:
        return np.nan, np.nan, np.nan

    return (
        float(x.median()),
        float(x.quantile(0.25)),
        float(x.quantile(0.75))
    )


print("Utilities ready.")
```

```python
from pathlib import Path
import json
import pandas as pd

pd.set_option("io.parquet.engine", "fastparquet")

BASE_DIR = Path.home() / "projects" / "diabetes-defer-lab-tests"

PROCESSED_DATA_DIR = BASE_DIR / "data" / "processed"
REPORT_DIR = BASE_DIR / "reports"

cohort_path = (
    PROCESSED_DATA_DIR
    / "step7_primary_modeling_cohort.parquet"
)

metadata_path = (
    REPORT_DIR
    / "step7_feature_metadata.json"
)

if not cohort_path.exists():
    raise FileNotFoundError(
        f"Missing cohort file: {cohort_path}"
    )

if not metadata_path.exists():
    raise FileNotFoundError(
        f"Missing metadata file: {metadata_path}"
    )

cohort = pd.read_parquet(
    cohort_path,
    engine="fastparquet"
)

with open(metadata_path, "r") as f:
    metadata = json.load(f)

TARGET_COL = metadata["target"]
cheap_features = metadata["cheap_features"]
lab_features = metadata["lab_features"]
full_features = metadata["full_features"]

print("Cohort loaded successfully.")
print("Shape:", cohort.shape)
print("Target:", TARGET_COL)
print("Cheap features:", len(cheap_features))
print("Lab features:", lab_features)
print("Splits:", cohort["split"].value_counts().to_dict())
```

```python
overall_summary = {
    "participants": int(len(cohort)),
    "target_positive": int(cohort[TARGET_COL].sum()),
    "target_prevalence": float(cohort[TARGET_COL].mean()),
    "train_n": int((cohort["split"] == "train").sum()),
    "validation_n": int((cohort["split"] == "validation").sum()),
    "test_n": int((cohort["split"] == "test").sum())
}

for lab in [
    "hba1c",
    "fasting_glucose",
    "two_hour_glucose"
]:
    if lab in cohort.columns:
        overall_summary[f"{lab}_availability"] = (
            float(cohort[lab].notna().mean())
        )

print(json.dumps(overall_summary, indent=2))

save_json(
    overall_summary,
    REPORT_DIR / "step8_overall_cohort_summary.json"
)
```

```python
def build_cohort_characteristics_table(df):
    rows = []

    numeric_variables = {
        "Age, years": "age",
        "BMI, kg/mÂ²": "bmi",
        "Waist circumference, cm": "waist_cm",
        "Systolic blood pressure, mmHg": "systolic_bp",
        "Diastolic blood pressure, mmHg": "diastolic_bp",
        "HbA1c, %": "hba1c",
        "Fasting glucose, mg/dL": "fasting_glucose",
        "Two-hour glucose, mg/dL": "two_hour_glucose"
    }

    for label, column in numeric_variables.items():
        if column not in df.columns:
            continue

        median, q1, q3 = median_iqr(df[column])

        rows.append({
            "Characteristic": label,
            "Value": (
                f"{median:.2f} "
                f"[{q1:.2f}, {q3:.2f}]"
                if pd.notna(median)
                else "NA"
            )
        })

    binary_variables = {
        "Screen-detected diabetes": TARGET_COL,
        "Overweight or obesity": "overweight_or_obesity",
        "Obesity": "obesity",
        "Hypertension-range BP": "hypertension_range_bp",
        "USPSTF-like screening eligibility": "uspstf_age_bmi_eligible"
    }

    for label, column in binary_variables.items():
        if column not in df.columns:
            continue

        valid = pd.to_numeric(
            df[column],
            errors="coerce"
        ).dropna()

        if len(valid) == 0:
            continue

        n_positive = int((valid == 1).sum())
        percentage = 100 * (valid == 1).mean()

        rows.append({
            "Characteristic": label,
            "Value": f"{n_positive:,} ({percentage:.1f}%)"
        })

    if "sex" in df.columns:
        for value, count in df["sex"].value_counts(dropna=True).items():
            percentage = (
                100 * count / df["sex"].notna().sum()
            )

            rows.append({
                "Characteristic": f"Sex: {value}",
                "Value": f"{count:,} ({percentage:.1f}%)"
            })

    return pd.DataFrame(rows)


cohort_table = build_cohort_characteristics_table(cohort)

display(cohort_table)

cohort_table.to_csv(
    TABLE_DIR / "step8_cohort_characteristics.csv",
    index=False
)
```

```python
split_prevalence = (
    cohort
    .groupby("split", observed=True)
    .agg(
        n=("SEQN", "count"),
        positives=(TARGET_COL, "sum"),
        prevalence=(TARGET_COL, "mean")
    )
    .reset_index()
)

split_order = [
    "train",
    "validation",
    "test"
]

split_prevalence["split"] = pd.Categorical(
    split_prevalence["split"],
    categories=split_order,
    ordered=True
)

split_prevalence = (
    split_prevalence
    .sort_values("split")
)

display(split_prevalence)

split_prevalence.to_csv(
    TABLE_DIR / "step8_target_prevalence_by_split.csv",
    index=False
)

fig = px.bar(
    split_prevalence,
    x="split",
    y="prevalence",
    text="prevalence",
    title="Screen-Detected Diabetes Prevalence by Temporal Split",
    color_discrete_sequence=[PROJECT_COLORS["teal_dark"]]
)

fig.update_traces(
    texttemplate="%{text:.1%}",
    textposition="outside"
)

fig.update_yaxes(
    tickformat=".0%",
    title="Prevalence"
)

fig.update_xaxes(title="Dataset split")

fig.show()

save_plotly(
    fig,
    "step8_target_prevalence_by_split"
)
```

```python
cycle_prevalence = (
    cohort
    .groupby("cycle", observed=True)
    .agg(
        n=("SEQN", "count"),
        positives=(TARGET_COL, "sum"),
        prevalence=(TARGET_COL, "mean")
    )
    .reset_index()
    .sort_values("cycle")
)

display(cycle_prevalence)

cycle_prevalence.to_csv(
    TABLE_DIR / "step8_target_prevalence_by_cycle.csv",
    index=False
)

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=cycle_prevalence["cycle"],
        y=cycle_prevalence["prevalence"],
        mode="lines+markers",
        line=dict(
            color=PROJECT_COLORS["purple_dark"],
            width=3
        ),
        marker=dict(size=9),
        name="Prevalence"
    )
)

fig.update_layout(
    title="Screen-Detected Diabetes Prevalence Across NHANES Cycles",
    xaxis_title="NHANES cycle",
    yaxis_title="Prevalence"
)

fig.update_yaxes(tickformat=".0%")

fig.show()

save_plotly(
    fig,
    "step8_target_prevalence_by_cycle"
)
```

```python
distribution_variables = {
    "age": "Age (years)",
    "bmi": "BMI (kg/mÂ²)",
    "waist_cm": "Waist circumference (cm)",
    "systolic_bp": "Systolic blood pressure (mmHg)",
    "diastolic_bp": "Diastolic blood pressure (mmHg)"
}

for column, label in distribution_variables.items():
    if column not in cohort.columns:
        continue

    plot_df = cohort[
        [column, TARGET_COL]
    ].dropna().copy()

    plot_df["Outcome"] = plot_df[TARGET_COL].map({
        0: "No screen-detected diabetes",
        1: "Screen-detected diabetes"
    })

    fig = px.histogram(
        plot_df,
        x=column,
        color="Outcome",
        marginal="box",
        histnorm="probability density",
        opacity=0.65,
        barmode="overlay",
        title=f"{label} by Diabetes Screening Outcome",
        color_discrete_sequence=[
            PROJECT_COLORS["teal"],
            PROJECT_COLORS["purple_dark"]
        ]
    )

    fig.update_xaxes(title=label)
    fig.update_yaxes(title="Density")

    fig.show()

    save_plotly(
        fig,
        f"step8_distribution_{column}"
    )
```

```python
LAB_THRESHOLDS = {
    "hba1c": {
        "label": "HbA1c (%)",
        "thresholds": [5.7, 6.5]
    },
    "fasting_glucose": {
        "label": "Fasting glucose (mg/dL)",
        "thresholds": [100, 126]
    },
    "two_hour_glucose": {
        "label": "Two-hour glucose (mg/dL)",
        "thresholds": [140, 200]
    }
}

for column, settings in LAB_THRESHOLDS.items():

    if column not in cohort.columns:
        continue

    lab_df = cohort[
        [column, TARGET_COL]
    ].dropna().copy()

    if len(lab_df) == 0:
        continue

    fig = px.histogram(
        lab_df,
        x=column,
        color=TARGET_COL,
        nbins=60,
        opacity=0.65,
        barmode="overlay",
        histnorm="probability density",
        title=f"{settings['label']} Distribution",
        color_discrete_sequence=[
            PROJECT_COLORS["green"],
            PROJECT_COLORS["purple_dark"]
        ]
    )

    for threshold in settings["thresholds"]:
        fig.add_vline(
            x=threshold,
            line_dash="dash",
            line_width=2,
            line_color=PROJECT_COLORS["yellow_dark"]
        )

    fig.update_xaxes(title=settings["label"])
    fig.update_yaxes(title="Density")

    fig.show()

    save_plotly(
        fig,
        f"step8_lab_distribution_{column}"
    )
```

```python
analysis_features = list(
    dict.fromkeys(
        cheap_features + lab_features
    )
)

missingness_records = []

for feature in analysis_features:
    if feature not in cohort.columns:
        continue

    missingness_records.append({
        "feature": feature,
        "missing_rate": float(
            cohort[feature].isna().mean()
        ),
        "available_rate": float(
            cohort[feature].notna().mean()
        ),
        "n_missing": int(
            cohort[feature].isna().sum()
        ),
        "n_available": int(
            cohort[feature].notna().sum()
        ),
        "tier": (
            "Lab"
            if feature in lab_features
            else "Cheap"
        )
    })

missingness_table = (
    pd.DataFrame(missingness_records)
    .sort_values(
        "missing_rate",
        ascending=False
    )
)

display(missingness_table)

missingness_table.to_csv(
    TABLE_DIR / "step8_feature_missingness.csv",
    index=False
)

fig = px.bar(
    missingness_table,
    x="missing_rate",
    y="feature",
    orientation="h",
    color="tier",
    title="Feature Missingness in the Primary Modeling Cohort",
    color_discrete_map={
        "Cheap": PROJECT_COLORS["teal_dark"],
        "Lab": PROJECT_COLORS["purple_dark"]
    }
)

fig.update_xaxes(
    tickformat=".0%",
    title="Missing rate"
)

fig.update_yaxes(title="Feature")

fig.update_layout(
    height=max(
        600,
        25 * len(missingness_table)
    )
)

fig.show()

save_plotly(
    fig,
    "step8_feature_missingness",
    height=max(
        700,
        30 * len(missingness_table)
    )
)
```

```python
SUBGROUP_CANDIDATES = [
    "sex",
    "age_group",
    "bmi_group",
    "race_ethnicity_code",
    "uspstf_age_bmi_eligible",
    "overweight_or_obesity",
    "obesity",
    "hypertension_range_bp",
    "cycle"
]


def summarize_subgroup_prevalence(
    df,
    group_col,
    target_col
):
    tmp = df[
        [group_col, target_col]
    ].dropna().copy()

    if len(tmp) == 0:
        return pd.DataFrame()

    result = (
        tmp
        .groupby(group_col, observed=True)
        .agg(
            n=(target_col, "size"),
            positives=(target_col, "sum"),
            prevalence=(target_col, "mean")
        )
        .reset_index()
    )

    result["group_variable"] = group_col

    return result


subgroup_tables = []

for subgroup in SUBGROUP_CANDIDATES:

    if subgroup not in cohort.columns:
        continue

    result = summarize_subgroup_prevalence(
        cohort,
        subgroup,
        TARGET_COL
    )

    if result.empty:
        continue

    subgroup_tables.append(result)

    result.to_csv(
        TABLE_DIR / f"step8_prevalence_by_{subgroup}.csv",
        index=False
    )

    plot_df = result.copy()
    plot_df[subgroup] = plot_df[subgroup].astype(str)

    fig = px.bar(
        plot_df,
        x=subgroup,
        y="prevalence",
        text="prevalence",
        title=f"Screen-Detected Diabetes Prevalence by {subgroup}",
        color_discrete_sequence=[
            PROJECT_COLORS["teal_dark"]
        ]
    )

    fig.update_traces(
        texttemplate="%{text:.1%}",
        textposition="outside"
    )

    fig.update_yaxes(
        tickformat=".0%",
        title="Prevalence"
    )

    fig.show()

    save_plotly(
        fig,
        f"step8_prevalence_by_{subgroup}"
    )
```

```python
lab_availability_records = []

for cycle, cycle_df in cohort.groupby(
    "cycle",
    observed=True
):

    for lab in lab_features:

        if lab not in cycle_df.columns:
            continue

        lab_availability_records.append({
            "cycle": cycle,
            "lab": lab,
            "availability_rate": float(
                cycle_df[lab].notna().mean()
            ),
            "available_n": int(
                cycle_df[lab].notna().sum()
            ),
            "total_n": int(
                len(cycle_df)
            )
        })

lab_availability = pd.DataFrame(
    lab_availability_records
)

display(lab_availability)

lab_availability.to_csv(
    TABLE_DIR / "step8_lab_availability_by_cycle.csv",
    index=False
)

fig = px.bar(
    lab_availability,
    x="cycle",
    y="availability_rate",
    color="lab",
    barmode="group",
    title="Laboratory Measurement Availability by NHANES Cycle",
    color_discrete_sequence=[
        PROJECT_COLORS["green_dark"],
        PROJECT_COLORS["yellow"],
        PROJECT_COLORS["purple_dark"]
    ]
)

fig.update_yaxes(
    tickformat=".0%",
    title="Availability rate"
)

fig.update_xaxes(
    title="NHANES cycle"
)

fig.show()

save_plotly(
    fig,
    "step8_lab_availability_by_cycle"
)
```

```python
radar_groups = []

if "sex" in cohort.columns:

    for group_name, group_df in cohort.groupby(
        "sex",
        observed=True
    ):

        if len(group_df) < 50:
            continue

        radar_groups.append({
            "group": str(group_name),
            "disease_prevalence": float(
                group_df[TARGET_COL].mean()
            ),
            "mean_age": float(
                pd.to_numeric(
                    group_df["age"],
                    errors="coerce"
                ).mean()
            ),
            "mean_bmi": float(
                pd.to_numeric(
                    group_df["bmi"],
                    errors="coerce"
                ).mean()
            ),
            "mean_systolic_bp": float(
                pd.to_numeric(
                    group_df["systolic_bp"],
                    errors="coerce"
                ).mean()
            ),
            "lab_availability": float(
                group_df[
                    ["hba1c", "fasting_glucose"]
                ].notna().any(axis=1).mean()
            )
        })

radar_df = pd.DataFrame(radar_groups)

if len(radar_df) > 0:

    metric_columns = [
        "disease_prevalence",
        "mean_age",
        "mean_bmi",
        "mean_systolic_bp",
        "lab_availability"
    ]

    radar_normalized = radar_df.copy()

    for column in metric_columns:
        minimum = radar_normalized[column].min()
        maximum = radar_normalized[column].max()

        if maximum > minimum:
            radar_normalized[column] = (
                radar_normalized[column] - minimum
            ) / (
                maximum - minimum
            )
        else:
            radar_normalized[column] = 0.5

    radar_normalized.to_csv(
        TABLE_DIR / "step8_screening_burden_radar.csv",
        index=False
    )

    fig = go.Figure()

    radar_colors = [
        PROJECT_COLORS["teal_dark"],
        PROJECT_COLORS["purple_dark"],
        PROJECT_COLORS["green_dark"],
        PROJECT_COLORS["yellow_dark"]
    ]

    for idx, row in radar_normalized.iterrows():

        values = [
            row[column]
            for column in metric_columns
        ]

        values.append(values[0])

        theta = metric_columns + [metric_columns[0]]

        fig.add_trace(
            go.Scatterpolar(
                r=values,
                theta=theta,
                fill="toself",
                name=row["group"],
                line=dict(
                    color=radar_colors[
                        idx % len(radar_colors)
                    ]
                ),
                opacity=0.55
            )
        )

    fig.update_layout(
        title="Relative Screening Burden Profile",
        polar=dict(
            radialaxis=dict(
                visible=True,
                range=[0, 1]
            )
        ),
        height=700
    )

    fig.show()

    save_plotly(
        fig,
        "step8_screening_burden_radar",
        height=700
    )
```

```python
def create_eda_narrative(
    cohort,
    target_col,
    split_prevalence
):

    n = len(cohort)
    positives = int(
        cohort[target_col].sum()
    )

    prevalence = float(
        cohort[target_col].mean()
    )

    lines = [
        "# Step 8 EDA Summary",
        "",
        f"The primary modeling cohort contains {n:,} adult NHANES participants.",
        "",
        (
            f"{positives:,} participants meet the primary "
            f"screen-detected diabetes definition, corresponding "
            f"to a prevalence of {prevalence:.2%}."
        ),
        "",
        "The study uses a temporal split:",
        ""
    ]

    for _, row in split_prevalence.iterrows():

        lines.append(
            f"- {row['split']}: {int(row['n']):,} participants, "
            f"{row['prevalence']:.2%} target prevalence."
        )

    lines.extend([
        "",
        (
            "The cheap-feature tier contains demographic, "
            "anthropometric, blood-pressure, lifestyle, and "
            "non-diabetes medical-history variables."
        ),
        "",
        (
            "HbA1c and fasting glucose constitute the primary "
            "expensive diagnostic laboratory measurements. "
            "Two-hour glucose is retained where available as an "
            "optional secondary laboratory feature."
        ),
        "",
        (
            "These descriptive results establish the population, "
            "missingness structure, laboratory coverage, subgroup "
            "heterogeneity, and temporal distribution shift that "
            "the predictive models must handle."
        )
    ])

    return "\n".join(lines)


eda_narrative = create_eda_narrative(
    cohort,
    TARGET_COL,
    split_prevalence
)

narrative_path = (
    REPORT_DIR
    / "step8_eda_narrative.md"
)

with open(narrative_path, "w") as f:
    f.write(eda_narrative)

print(eda_narrative)
print("\nSaved:", narrative_path)
```

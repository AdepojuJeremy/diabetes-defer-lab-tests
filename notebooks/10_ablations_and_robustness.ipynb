```python
class MLPBlock(nn.Module):

    def __init__(
        self,
        input_dim,
        hidden_dim=128,
        num_layers=3,
        dropout=0.15
    ):
        super().__init__()

        layers = []

        current_dim = input_dim

        for _ in range(
            num_layers
        ):

            layers.extend([
                nn.Linear(
                    current_dim,
                    hidden_dim
                ),

                nn.LayerNorm(
                    hidden_dim
                ),

                nn.GELU(),

                nn.Dropout(
                    dropout
                )
            ])

            current_dim = (
                hidden_dim
            )

        self.network = (
            nn.Sequential(
                *layers
            )
        )

    def forward(
        self,
        x
    ):
        return self.network(
            x
        )


class DeferToLabNet(nn.Module):

    def __init__(
        self,
        cheap_input_dim,
        full_input_dim,
        hidden_dim=128,
        num_layers=3,
        dropout=0.15
    ):
        super().__init__()

        self.cheap_encoder = MLPBlock(
            cheap_input_dim,
            hidden_dim,
            num_layers,
            dropout
        )

        self.full_encoder = MLPBlock(
            full_input_dim,
            hidden_dim,
            num_layers,
            dropout
        )

        self.gate_encoder = MLPBlock(
            cheap_input_dim,
            hidden_dim,
            2,
            dropout
        )

        self.cheap_head = nn.Linear(
            hidden_dim,
            1
        )

        self.full_head = nn.Linear(
            hidden_dim,
            1
        )

        self.defer_head = nn.Linear(
            hidden_dim,
            1
        )

    def forward(
        self,
        x_cheap,
        x_full
    ):

        cheap_rep = (
            self.cheap_encoder(
                x_cheap
            )
        )

        full_rep = (
            self.full_encoder(
                x_full
            )
        )

        gate_rep = (
            self.gate_encoder(
                x_cheap
            )
        )

        return {
            "cheap_logit": (
                self.cheap_head(
                    cheap_rep
                )
                .squeeze(-1)
            ),

            "full_logit": (
                self.full_head(
                    full_rep
                )
                .squeeze(-1)
            ),

            "defer_logit": (
                self.defer_head(
                    gate_rep
                )
                .squeeze(-1)
            )
        }
```

```python
import numpy as np
import pandas as pd
import torch

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    f1_score,
    confusion_matrix
)


def safe_divide(numerator, denominator):
    if denominator == 0:
        return np.nan

    return numerator / denominator


@torch.no_grad()
def predict_model(
    model,
    X_cheap,
    X_full,
    batch_size=2048
):

    model.eval()

    cheap_probabilities = []
    full_probabilities = []
    defer_probabilities = []

    for start in range(
        0,
        len(X_cheap),
        batch_size
    ):

        end = min(
            start + batch_size,
            len(X_cheap)
        )

        xc = torch.tensor(
            X_cheap[start:end],
            dtype=torch.float32,
            device=DEVICE
        )

        xf = torch.tensor(
            X_full[start:end],
            dtype=torch.float32,
            device=DEVICE
        )

        outputs = model(
            xc,
            xf
        )

        cheap_probabilities.append(
            torch.sigmoid(
                outputs["cheap_logit"]
            )
            .float()
            .cpu()
            .numpy()
        )

        full_probabilities.append(
            torch.sigmoid(
                outputs["full_logit"]
            )
            .float()
            .cpu()
            .numpy()
        )

        defer_probabilities.append(
            torch.sigmoid(
                outputs["defer_logit"]
            )
            .float()
            .cpu()
            .numpy()
        )

    return {
        "cheap_probability": (
            np.concatenate(
                cheap_probabilities
            )
            .astype(np.float32)
        ),

        "full_probability": (
            np.concatenate(
                full_probabilities
            )
            .astype(np.float32)
        ),

        "defer_probability": (
            np.concatenate(
                defer_probabilities
            )
            .astype(np.float32)
        )
    }


def threshold_for_budget(
    scores,
    budget
):

    scores = np.asarray(
        scores,
        dtype=np.float32
    )

    budget = float(
        np.clip(
            budget,
            0.0,
            1.0
        )
    )

    if budget <= 0:
        return np.inf

    if budget >= 1:
        return -np.inf

    try:
        threshold = np.quantile(
            scores,
            1.0 - budget,
            method="linear"
        )

    except TypeError:
        threshold = np.quantile(
            scores,
            1.0 - budget,
            interpolation="linear"
        )

    return float(
        threshold
    )


def laboratory_rule(df):

    hba1c = pd.to_numeric(
        df["hba1c"],
        errors="coerce"
    )

    fasting_glucose = pd.to_numeric(
        df["fasting_glucose"],
        errors="coerce"
    )

    return (
        (hba1c >= 6.5)
        |
        (fasting_glucose >= 126)
    ).astype(
        int
    ).to_numpy()


def evaluate_policy(
    y_true,
    cheap_probability,
    referral_mask,
    lab_prediction,
    cheap_threshold,
    lab_cost=1.0,
    false_negative_cost=10.0,
    false_positive_cost=2.0
):

    y_true = np.asarray(
        y_true
    ).astype(int)

    cheap_probability = np.asarray(
        cheap_probability
    )

    referral_mask = np.asarray(
        referral_mask
    ).astype(bool)

    lab_prediction = np.asarray(
        lab_prediction
    ).astype(int)

    cheap_prediction = (
        cheap_probability
        >= cheap_threshold
    ).astype(int)

    final_prediction = (
        cheap_prediction.copy()
    )

    final_prediction[
        referral_mask
    ] = lab_prediction[
        referral_mask
    ]

    tn, fp, fn, tp = (
        confusion_matrix(
            y_true,
            final_prediction,
            labels=[0, 1]
        )
        .ravel()
    )

    sensitivity = safe_divide(
        tp,
        tp + fn
    )

    specificity = safe_divide(
        tn,
        tn + fp
    )

    referral_rate = float(
        referral_mask.mean()
    )

    lab_cost_component = (
        lab_cost
        * referral_rate
    )

    false_negative_cost_component = (
        false_negative_cost
        * fn
        / len(y_true)
    )

    false_positive_cost_component = (
        false_positive_cost
        * fp
        / len(y_true)
    )

    total_normalized_cost = (
        lab_cost_component
        + false_negative_cost_component
        + false_positive_cost_component
    )

    return {
        "n": int(
            len(y_true)
        ),

        "referral_rate": (
            referral_rate
        ),

        "accuracy": (
            accuracy_score(
                y_true,
                final_prediction
            )
        ),

        "balanced_accuracy": (
            balanced_accuracy_score(
                y_true,
                final_prediction
            )
        ),

        "sensitivity": (
            sensitivity
        ),

        "specificity": (
            specificity
        ),

        "precision": (
            precision_score(
                y_true,
                final_prediction,
                zero_division=0
            )
        ),

        "f1": (
            f1_score(
                y_true,
                final_prediction,
                zero_division=0
            )
        ),

        "tp": int(tp),
        "tn": int(tn),
        "fp": int(fp),
        "fn": int(fn),

        "missed_case_rate": (
            1.0 - sensitivity
        ),

        "lab_cost_component": (
            lab_cost_component
        ),

        "false_negative_cost_component": (
            false_negative_cost_component
        ),

        "false_positive_cost_component": (
            false_positive_cost_component
        ),

        "total_normalized_cost": (
            total_normalized_cost
        )
    }


print("Step 14 prediction and policy utilities restored.")
```

```python
class DeferDataset(Dataset):

    def __init__(
        self,
        X_cheap,
        X_full,
        y
    ):

        self.X_cheap = torch.tensor(
            X_cheap,
            dtype=torch.float32
        )

        self.X_full = torch.tensor(
            X_full,
            dtype=torch.float32
        )

        self.y = torch.tensor(
            y,
            dtype=torch.float32
        )

    def __len__(
        self
    ):
        return len(
            self.y
        )

    def __getitem__(
        self,
        index
    ):

        return (
            self.X_cheap[
                index
            ],

            self.X_full[
                index
            ],

            self.y[
                index
            ]
        )


train_dataset = DeferDataset(
    X_train_cheap,
    X_train_full,
    y_train
)

val_dataset = DeferDataset(
    X_val_cheap,
    X_val_full,
    y_val
)

train_loader = DataLoader(
    train_dataset,
    batch_size=(
        RETRAIN_BATCH_SIZE
    ),
    shuffle=True,
    num_workers=0,
    pin_memory=USE_AMP
)

val_loader = DataLoader(
    val_dataset,
    batch_size=(
        RETRAIN_BATCH_SIZE
    ),
    shuffle=False,
    num_workers=0,
    pin_memory=USE_AMP
)
```

```python
def budget_specific_loss(
    outputs,
    targets,
    target_budget,
    defer_cost_lambda=0.10,
    budget_penalty_lambda=10.0
):

    cheap_logit = (
        outputs[
            "cheap_logit"
        ]
    )

    full_logit = (
        outputs[
            "full_logit"
        ]
    )

    defer_probability = (
        torch.sigmoid(
            outputs[
                "defer_logit"
            ]
        )
    )

    cheap_loss = (
        F.binary_cross_entropy_with_logits(
            cheap_logit,
            targets,
            pos_weight=(
                POS_WEIGHT_TENSOR
            ),
            reduction="none"
        )
    )

    full_loss = (
        F.binary_cross_entropy_with_logits(
            full_logit,
            targets,
            pos_weight=(
                POS_WEIGHT_TENSOR
            ),
            reduction="none"
        )
    )

    selective_loss = (
        (
            1.0
            - defer_probability
        )
        * cheap_loss
        +
        defer_probability
        * full_loss
    ).mean()

    referral_rate = (
        defer_probability.mean()
    )

    cost_penalty = (
        defer_cost_lambda
        * referral_rate
    )

    budget_violation = torch.relu(
        referral_rate
        - target_budget
    )

    budget_penalty = (
        budget_penalty_lambda
        * budget_violation.pow(2)
    )

    total_loss = (
        selective_loss
        + cost_penalty
        + budget_penalty
    )

    return (
        total_loss,
        referral_rate
    )
```

```python
def train_model_for_budget(
    target_budget,
    seed=42
):

    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(
            seed
        )

    candidate_model = (
        DeferToLabNet(
            cheap_input_dim=(
                X_train_cheap.shape[1]
            ),

            full_input_dim=(
                X_train_full.shape[1]
            ),

            hidden_dim=128,
            num_layers=3,
            dropout=0.15
        )
        .to(
            DEVICE
        )
    )

    optimizer = torch.optim.AdamW(
        candidate_model.parameters(),
        lr=RETRAIN_LEARNING_RATE,
        weight_decay=(
            RETRAIN_WEIGHT_DECAY
        )
    )

    scaler = torch.cuda.amp.GradScaler(
        enabled=USE_AMP
    )

    best_validation_loss = np.inf

    best_state = None

    best_epoch = 0

    patience_counter = 0

    history = []

    for epoch in range(
        1,
        RETRAIN_MAX_EPOCHS + 1
    ):

        candidate_model.train()

        epoch_loss = 0.0

        epoch_n = 0

        for (
            x_cheap,
            x_full,
            targets
        ) in train_loader:

            x_cheap = x_cheap.to(
                DEVICE
            )

            x_full = x_full.to(
                DEVICE
            )

            targets = targets.to(
                DEVICE
            )

            optimizer.zero_grad(
                set_to_none=True
            )

            with torch.cuda.amp.autocast(
                enabled=USE_AMP
            ):

                outputs = (
                    candidate_model(
                        x_cheap,
                        x_full
                    )
                )

                loss, referral_rate = (
                    budget_specific_loss(
                        outputs,
                        targets,
                        target_budget
                    )
                )

            scaler.scale(
                loss
            ).backward()

            scaler.unscale_(
                optimizer
            )

            torch.nn.utils.clip_grad_norm_(
                candidate_model.parameters(),
                5.0
            )

            scaler.step(
                optimizer
            )

            scaler.update()

            epoch_loss += (
                float(
                    loss.detach()
                    .cpu()
                )
                * len(
                    targets
                )
            )

            epoch_n += len(
                targets
            )

        train_loss = (
            epoch_loss
            / epoch_n
        )

        candidate_model.eval()

        validation_loss_total = 0.0

        validation_n = 0

        with torch.no_grad():

            for (
                x_cheap,
                x_full,
                targets
            ) in val_loader:

                x_cheap = x_cheap.to(
                    DEVICE
                )

                x_full = x_full.to(
                    DEVICE
                )

                targets = targets.to(
                    DEVICE
                )

                with torch.cuda.amp.autocast(
                    enabled=USE_AMP
                ):

                    outputs = (
                        candidate_model(
                            x_cheap,
                            x_full
                        )
                    )

                    loss, _ = (
                        budget_specific_loss(
                            outputs,
                            targets,
                            target_budget
                        )
                    )

                validation_loss_total += (
                    float(
                        loss.detach()
                        .cpu()
                    )
                    * len(
                        targets
                    )
                )

                validation_n += len(
                    targets
                )

        validation_loss = (
            validation_loss_total
            / validation_n
        )

        history.append({
            "epoch": (
                epoch
            ),

            "train_loss": (
                train_loss
            ),

            "validation_loss": (
                validation_loss
            )
        })

        if (
            validation_loss
            <
            best_validation_loss
            - 1e-4
        ):

            best_validation_loss = (
                validation_loss
            )

            best_epoch = (
                epoch
            )

            patience_counter = 0

            best_state = {
                key: value
                .detach()
                .cpu()
                .clone()

                for key, value
                in candidate_model
                .state_dict()
                .items()
            }

        else:

            patience_counter += 1

        if (
            epoch == 1
            or
            epoch % 5 == 0
        ):

            print(
                f"Budget {target_budget:.0%} | "
                f"Epoch {epoch:02d} | "
                f"Train {train_loss:.4f} | "
                f"Val {validation_loss:.4f}"
            )

        if (
            patience_counter
            >= RETRAIN_PATIENCE
        ):

            break

    candidate_model.load_state_dict(
        best_state
    )

    candidate_model.to(
        DEVICE
    )

    candidate_model.eval()

    return (
        candidate_model,
        best_epoch,
        best_validation_loss,
        pd.DataFrame(
            history
        )
    )
```

```python
budget_retraining_records = []


if RUN_SELECTED_BUDGET_RETRAINING:

    retraining_start = (
        time.time()
    )

    for budget in (
        RETRAIN_BUDGETS
    ):

        print()
        print(
            "=" * 50
        )

        print(
            f"Retraining for "
            f"{budget:.0%} laboratory budget"
        )

        candidate_model, best_epoch, best_val_loss, history = (
            train_model_for_budget(
                target_budget=budget,
                seed=SEED
            )
        )

        val_candidate_outputs = (
            predict_model(
                candidate_model,
                X_val_cheap,
                X_val_full
            )
        )

        test_candidate_outputs = (
            predict_model(
                candidate_model,
                X_test_cheap,
                X_test_full
            )
        )

        candidate_defer_threshold = (
            threshold_for_budget(
                val_candidate_outputs[
                    "defer_probability"
                ],
                budget
            )
        )

        candidate_referral_mask = (
            test_candidate_outputs[
                "defer_probability"
            ]
            >= candidate_defer_threshold
        )

        # Recalibrate cheap classifier threshold
        # on validation data for this retrained model.
        fpr, tpr, thresholds = roc_curve(
            y_val,
            val_candidate_outputs[
                "cheap_probability"
            ]
        )

        candidate_cheap_threshold = float(
            thresholds[
                np.nanargmax(
                    tpr - fpr
                )
            ]
        )

        metrics = evaluate_policy(
            y_true=y_test,

            cheap_probability=(
                test_candidate_outputs[
                    "cheap_probability"
                ]
            ),

            referral_mask=(
                candidate_referral_mask
            ),

            lab_prediction=(
                lab_rule_test
            ),

            cheap_threshold=(
                candidate_cheap_threshold
            )
        )

        budget_retraining_records.append({
            "training_budget": (
                budget
            ),

            "best_epoch": (
                best_epoch
            ),

            "best_validation_loss": (
                best_val_loss
            ),

            **metrics
        })

        model_path = (
            STEP14_MODEL_DIR
            / (
                f"budget_{int(budget * 100)}"
                "_robustness_model.pt"
            )
        )

        torch.save(
            {
                "model_state_dict": (
                    candidate_model
                    .state_dict()
                ),

                "training_budget": (
                    budget
                ),

                "best_epoch": (
                    best_epoch
                ),

                "cheap_threshold": (
                    candidate_cheap_threshold
                ),

                "defer_threshold": (
                    candidate_defer_threshold
                )
            },
            model_path
        )

        history.to_csv(
            TABLE_DIR
            / (
                f"step14_retraining_history_"
                f"budget_{int(budget * 100)}.csv"
            ),
            index=False
        )

        del candidate_model

        gc.collect()

        if torch.cuda.is_available():
            torch.cuda.empty_cache()

    retraining_minutes = (
        time.time()
        - retraining_start
    ) / 60

else:

    retraining_minutes = 0.0


budget_retraining = pd.DataFrame(
    budget_retraining_records
)

if len(
    budget_retraining
) > 0:

    display(
        budget_retraining
    )

    budget_retraining.to_csv(
        TABLE_DIR
        / "step14_selected_budget_retraining.csv",
        index=False
    )


print(
    "Selected retraining time:",
    round(
        retraining_minutes,
        2
    ),
    "minutes"
)
```

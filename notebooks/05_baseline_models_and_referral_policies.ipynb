```python
def make_one_hot_encoder():
    try:
        return OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        )
    except TypeError:
        return OneHotEncoder(
            handle_unknown="ignore",
            sparse=False
        )


def build_preprocessor(
    numeric_features,
    categorical_features
):
    transformers = []

    if numeric_features:
        numeric_pipeline = Pipeline([
            (
                "imputer",
                SimpleImputer(strategy="median")
            ),
            (
                "scaler",
                StandardScaler()
            )
        ])

        transformers.append(
            (
                "numeric",
                numeric_pipeline,
                numeric_features
            )
        )

    if categorical_features:
        categorical_pipeline = Pipeline([
            (
                "imputer",
                SimpleImputer(
                    strategy="most_frequent"
                )
            ),
            (
                "onehot",
                make_one_hot_encoder()
            )
        ])

        transformers.append(
            (
                "categorical",
                categorical_pipeline,
                categorical_features
            )
        )

    return ColumnTransformer(
        transformers=transformers,
        remainder="drop"
    )


cheap_preprocessor = build_preprocessor(
    cheap_numeric_features,
    cheap_categorical_features
)

full_preprocessor = build_preprocessor(
    primary_full_numeric_features,
    primary_full_categorical_features
)

print("Preprocessors ready.")
```

```python
def build_model_registry():
    return {
        "logistic_regression": LogisticRegression(
            max_iter=3000,
            class_weight="balanced",
            random_state=42
        ),

        "random_forest": RandomForestClassifier(
            n_estimators=500,
            max_depth=None,
            min_samples_leaf=4,
            max_features="sqrt",
            class_weight="balanced_subsample",
            n_jobs=-1,
            random_state=42
        ),

        "hist_gradient_boosting": HistGradientBoostingClassifier(
            learning_rate=0.05,
            max_iter=350,
            max_leaf_nodes=31,
            min_samples_leaf=20,
            l2_regularization=1e-3,
            class_weight="balanced",
            random_state=42
        )
    }


def build_pipeline(
    preprocessor,
    model
):
    return Pipeline([
        ("preprocessor", preprocessor),
        ("classifier", model)
    ])


print(build_model_registry())
```

```python
cheap_models = {}

start_time = time.time()

for model_name, classifier in build_model_registry().items():

    print(f"Training cheap-only {model_name}...")

    pipeline = build_pipeline(
        build_preprocessor(
            cheap_numeric_features,
            cheap_categorical_features
        ),
        classifier
    )

    pipeline.fit(
        X_train_cheap,
        y_train
    )

    cheap_models[model_name] = pipeline

    path = (
        STEP9_MODEL_DIR /
        f"cheap_{model_name}.joblib"
    )

    joblib.dump(
        pipeline,
        path
    )

    print("Saved:", path)

print(
    "Cheap model training time:",
    round((time.time() - start_time) / 60, 2),
    "minutes"
)
```

```python
def predict_probability(model, X):
    return model.predict_proba(X)[:, 1]


def calculate_binary_metrics(
    y_true,
    probability,
    threshold=0.5
):
    y_true = np.asarray(y_true).astype(int)
    probability = np.asarray(probability)

    prediction = (
        probability >= threshold
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        prediction,
        labels=[0, 1]
    ).ravel()

    sensitivity = (
        tp / (tp + fn)
        if (tp + fn) > 0
        else np.nan
    )

    specificity = (
        tn / (tn + fp)
        if (tn + fp) > 0
        else np.nan
    )

    ppv = (
        tp / (tp + fp)
        if (tp + fp) > 0
        else np.nan
    )

    npv = (
        tn / (tn + fn)
        if (tn + fn) > 0
        else np.nan
    )

    return {
        "auroc": roc_auc_score(
            y_true,
            probability
        ),

        "auprc": average_precision_score(
            y_true,
            probability
        ),

        "accuracy": accuracy_score(
            y_true,
            prediction
        ),

        "balanced_accuracy": balanced_accuracy_score(
            y_true,
            prediction
        ),

        "sensitivity": sensitivity,
        "specificity": specificity,
        "ppv": ppv,
        "npv": npv,

        "f1": f1_score(
            y_true,
            prediction,
            zero_division=0
        ),

        "brier": brier_score_loss(
            y_true,
            probability
        ),

        "threshold": float(threshold),

        "tp": int(tp),
        "tn": int(tn),
        "fp": int(fp),
        "fn": int(fn)
    }


def find_youden_threshold(
    y_true,
    probability
):
    fpr, tpr, thresholds = roc_curve(
        y_true,
        probability
    )

    youden = (
        tpr - fpr
    )

    idx = np.nanargmax(youden)

    return float(thresholds[idx])


def find_threshold_for_sensitivity(
    y_true,
    probability,
    target_sensitivity=0.90
):
    thresholds = np.linspace(
        0,
        1,
        1001
    )

    candidates = []

    for threshold in thresholds:

        prediction = (
            probability >= threshold
        ).astype(int)

        tn, fp, fn, tp = confusion_matrix(
            y_true,
            prediction,
            labels=[0, 1]
        ).ravel()

        sensitivity = (
            tp / (tp + fn)
            if (tp + fn) > 0
            else 0
        )

        specificity = (
            tn / (tn + fp)
            if (tn + fp) > 0
            else 0
        )

        if sensitivity >= target_sensitivity:

            candidates.append(
                (
                    threshold,
                    sensitivity,
                    specificity
                )
            )

    if not candidates:
        return 0.5

    best = max(
        candidates,
        key=lambda x: (
            x[2],
            x[0]
        )
    )

    return float(best[0])
```

```python
baseline_records = []
prediction_data = pd.DataFrame({
    "SEQN": test_df["SEQN"].values,
    "cycle": test_df["cycle"].values,
    "y_true": y_test.values
})

for tier_name, models, X_val, X_test in [
    (
        "cheap",
        cheap_models,
        X_val_cheap,
        X_test_cheap
    ),
    (
        "full_lab",
        full_models,
        X_val_full,
        X_test_full
    )
]:

    for model_name, model in models.items():

        val_prob = predict_probability(
            model,
            X_val
        )

        test_prob = predict_probability(
            model,
            X_test
        )

        youden_threshold = find_youden_threshold(
            y_val,
            val_prob
        )

        sensitivity90_threshold = (
            find_threshold_for_sensitivity(
                y_val,
                val_prob,
                target_sensitivity=0.90
            )
        )

        for threshold_name, threshold in [
            (
                "youden_validation",
                youden_threshold
            ),
            (
                "sensitivity90_validation",
                sensitivity90_threshold
            )
        ]:

            metrics = calculate_binary_metrics(
                y_test,
                test_prob,
                threshold=threshold
            )

            baseline_records.append({
                "tier": tier_name,
                "model": model_name,
                "threshold_strategy": threshold_name,
                **metrics
            })

        probability_column = (
            f"{tier_name}_{model_name}_prob"
        )

        prediction_data[
            probability_column
        ] = test_prob


baseline_metrics = pd.DataFrame(
    baseline_records
)

display(
    baseline_metrics.sort_values(
        "auroc",
        ascending=False
    )
)

baseline_metrics.to_csv(
    TABLE_DIR /
    "step9_baseline_model_metrics.csv",
    index=False
)

prediction_data.to_parquet(
    PROCESSED_DATA_DIR /
    "step9_baseline_predictions.parquet",
    engine="fastparquet",
    index=False
)
```

```python
def laboratory_diagnosis(df):
    hba1c_positive = (
        pd.to_numeric(
            df["hba1c"],
            errors="coerce"
        ) >= 6.5
    )

    glucose_positive = (
        pd.to_numeric(
            df["fasting_glucose"],
            errors="coerce"
        ) >= 126
    )

    return (
        hba1c_positive |
        glucose_positive
    ).astype(int)


lab_rule_val = laboratory_diagnosis(
    val_df
)

lab_rule_test = laboratory_diagnosis(
    test_df
)

print(
    "Validation lab-rule accuracy:",
    accuracy_score(
        y_val,
        lab_rule_val
    )
)

print(
    "Test lab-rule accuracy:",
    accuracy_score(
        y_test,
        lab_rule_test
    )
)
```

```python
LAB_REFERRAL_COST = 1.0
FALSE_NEGATIVE_COST = 10.0
FALSE_POSITIVE_COST = 2.0


def evaluate_referral_policy(
    y_true,
    cheap_probability,
    cheap_threshold,
    referral_mask,
    lab_prediction
):
    y_true = np.asarray(
        y_true
    ).astype(int)

    cheap_probability = np.asarray(
        cheap_probability
    )

    referral_mask = np.asarray(
        referral_mask
    ).astype(bool)

    lab_prediction = np.asarray(
        lab_prediction
    ).astype(int)

    cheap_prediction = (
        cheap_probability >=
        cheap_threshold
    ).astype(int)

    final_prediction = (
        cheap_prediction.copy()
    )

    final_prediction[
        referral_mask
    ] = lab_prediction[
        referral_mask
    ]

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        final_prediction,
        labels=[0, 1]
    ).ravel()

    sensitivity = (
        tp / (tp + fn)
        if (tp + fn) > 0
        else np.nan
    )

    specificity = (
        tn / (tn + fp)
        if (tn + fp) > 0
        else np.nan
    )

    referral_rate = float(
        referral_mask.mean()
    )

    average_lab_cost = (
        referral_rate *
        LAB_REFERRAL_COST
    )

    misclassification_cost = (
        FALSE_NEGATIVE_COST * fn +
        FALSE_POSITIVE_COST * fp
    ) / len(y_true)

    total_normalized_cost = (
        average_lab_cost +
        misclassification_cost
    )

    return {
        "accuracy": accuracy_score(
            y_true,
            final_prediction
        ),

        "balanced_accuracy": (
            balanced_accuracy_score(
                y_true,
                final_prediction
            )
        ),

        "sensitivity": sensitivity,
        "specificity": specificity,

        "f1": f1_score(
            y_true,
            final_prediction,
            zero_division=0
        ),

        "referral_rate": referral_rate,
        "average_lab_cost": average_lab_cost,
        "misclassification_cost": (
            misclassification_cost
        ),
        "total_normalized_cost": (
            total_normalized_cost
        ),

        "tp": int(tp),
        "tn": int(tn),
        "fp": int(fp),
        "fn": int(fn)
    }
```

```python
risk_threshold_records = []

for referral_threshold in np.linspace(
    0.05,
    0.95,
    91
):

    referral_mask = (
        cheap_val_prob >=
        referral_threshold
    )

    metrics = evaluate_referral_policy(
        y_true=y_val,
        cheap_probability=cheap_val_prob,
        cheap_threshold=cheap_classification_threshold,
        referral_mask=referral_mask,
        lab_prediction=lab_rule_val
    )

    risk_threshold_records.append({
        "referral_threshold": float(
            referral_threshold
        ),
        **metrics
    })


risk_tuning = pd.DataFrame(
    risk_threshold_records
)

best_risk_row = (
    risk_tuning
    .sort_values(
        [
            "total_normalized_cost",
            "sensitivity"
        ],
        ascending=[
            True,
            False
        ]
    )
    .iloc[0]
)

BEST_RISK_REFERRAL_THRESHOLD = float(
    best_risk_row[
        "referral_threshold"
    ]
)

print(
    "Best risk referral threshold:",
    BEST_RISK_REFERRAL_THRESHOLD
)

display(
    risk_tuning.sort_values(
        "total_normalized_cost"
    ).head(10)
)

risk_tuning.to_csv(
    TABLE_DIR /
    "step9_risk_referral_validation_tuning.csv",
    index=False
)
```

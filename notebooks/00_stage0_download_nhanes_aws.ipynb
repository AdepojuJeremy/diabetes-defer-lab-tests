```python
NHANES_CYCLES = {
    "2011": {"cycle": "2011-2012", "suffix": "G"},
    "2013": {"cycle": "2013-2014", "suffix": "H"},
    "2015": {"cycle": "2015-2016", "suffix": "I"},
    "2017": {"cycle": "2017-2018", "suffix": "J"}
}

CORE_MODULES = {
    "DEMO": "Demographics",
    "BMX": "Body measures",
    "BPX": "Blood pressure examination",
    "BPQ": "Blood pressure questionnaire",
    "DIQ": "Diabetes questionnaire",
    "GHB": "Glycohemoglobin / HbA1c",
    "GLU": "Fasting glucose",
    "OGTT": "Oral glucose tolerance test",
    "SMQ": "Smoking questionnaire",
    "PAQ": "Physical activity questionnaire"
}

OPTIONAL_MODULES = {
    "ALQ": "Alcohol questionnaire",
    "MCQ": "Medical conditions questionnaire"
}

MODULES_TO_DOWNLOAD = {
    **CORE_MODULES,
    **OPTIONAL_MODULES
}

def build_nhanes_download_plan(cycles, modules):
    records = []

    for year_key, cycle_info in cycles.items():
        suffix = cycle_info["suffix"]

        for module, description in modules.items():
            filename = f"{module}_{suffix}.XPT"
            doc_filename = f"{module}_{suffix}.htm"

            records.append({
                "year_key": year_key,
                "cycle": cycle_info["cycle"],
                "suffix": suffix,
                "module": module,
                "description": description,
                "xpt_filename": filename,
                "doc_filename": doc_filename,
                "xpt_url": f"https://wwwn.cdc.gov/Nchs/Data/Nhanes/Public/{year_key}/DataFiles/{filename}",
                "doc_url": f"https://wwwn.cdc.gov/Nchs/Data/Nhanes/Public/{year_key}/DataFiles/{doc_filename}",
                "is_core": module in CORE_MODULES
            })

    return pd.DataFrame(records)

download_plan = build_nhanes_download_plan(
    NHANES_CYCLES,
    MODULES_TO_DOWNLOAD
)

download_plan_path = REGISTRY_DIR / "stage0_nhanes_download_plan.csv"
download_plan.to_csv(download_plan_path, index=False)

display(download_plan)

print(f"Planned files: {len(download_plan)}")
print(f"Download plan saved to: {download_plan_path}")
```

```python
def file_sha256(path, chunk_size=1024 * 1024):
    hasher = hashlib.sha256()

    with open(path, "rb") as f:
        while True:
            chunk = f.read(chunk_size)

            if not chunk:
                break

            hasher.update(chunk)

    return hasher.hexdigest()


def download_file(
    url,
    destination,
    min_bytes=1024,
    retries=3,
    timeout=60
):
    destination = Path(destination)
    destination.parent.mkdir(parents=True, exist_ok=True)

    if destination.exists() and destination.stat().st_size >= min_bytes:
        return {
            "status": "exists",
            "path": str(destination),
            "bytes": int(destination.stat().st_size),
            "sha256": file_sha256(destination)
        }

    temp_path = destination.with_suffix(destination.suffix + ".part")

    last_error = None

    for attempt in range(1, retries + 1):
        try:
            with requests.get(url, stream=True, timeout=timeout) as response:
                response.raise_for_status()

                total = int(response.headers.get("content-length", 0))

                with open(temp_path, "wb") as f:
                    with tqdm(
                        total=total,
                        unit="B",
                        unit_scale=True,
                        desc=destination.name,
                        leave=False
                    ) as progress:
                        for chunk in response.iter_content(chunk_size=1024 * 1024):
                            if chunk:
                                f.write(chunk)
                                progress.update(len(chunk))

            if temp_path.stat().st_size < min_bytes:
                raise ValueError(
                    f"Downloaded file too small: {temp_path.stat().st_size} bytes"
                )

            temp_path.replace(destination)

            return {
                "status": "downloaded",
                "path": str(destination),
                "bytes": int(destination.stat().st_size),
                "sha256": file_sha256(destination)
            }

        except Exception as error:
            last_error = str(error)

            if temp_path.exists():
                temp_path.unlink()

            time.sleep(3)

    return {
        "status": "failed",
        "path": str(destination),
        "bytes": 0,
        "sha256": None,
        "error": last_error
    }

print("Downloader ready.")
```

```python
xpt_records = []

for row in download_plan.itertuples(index=False):
    destination = RAW_DATA_DIR / row.xpt_filename

    result = download_file(
        url=row.xpt_url,
        destination=destination,
        min_bytes=2048,
        retries=3,
        timeout=90
    )

    record = {
        "cycle": row.cycle,
        "year_key": row.year_key,
        "module": row.module,
        "description": row.description,
        "filename": row.xpt_filename,
        "url": row.xpt_url,
        "is_core": row.is_core,
        **result
    }

    xpt_records.append(record)

xpt_download_registry = pd.DataFrame(xpt_records)

xpt_registry_path = REGISTRY_DIR / "stage0_xpt_download_registry.csv"
xpt_download_registry.to_csv(xpt_registry_path, index=False)

display(xpt_download_registry)

print(f"XPT download registry saved to: {xpt_registry_path}")

failed_xpt = xpt_download_registry[xpt_download_registry["status"] == "failed"]

if len(failed_xpt) > 0:
    print("Some files failed. We can retry them later.")
    display(failed_xpt[["cycle", "module", "filename", "url", "error"]])
else:
    print("All planned XPT files downloaded or already existed.")
```

```python
def verify_stage0_downloads(download_plan, raw_data_dir):
    records = []

    for row in download_plan.itertuples(index=False):
        path = raw_data_dir / row.xpt_filename

        records.append({
            "cycle": row.cycle,
            "module": row.module,
            "filename": row.xpt_filename,
            "is_core": row.is_core,
            "exists": path.exists(),
            "bytes": int(path.stat().st_size) if path.exists() else 0,
            "path": str(path)
        })

    return pd.DataFrame(records)

download_summary = verify_stage0_downloads(
    download_plan,
    RAW_DATA_DIR
)

summary_path = REGISTRY_DIR / "stage0_download_summary.csv"
download_summary.to_csv(summary_path, index=False)

missing_core_files = download_summary[
    (download_summary["is_core"])
    & (~download_summary["exists"])
].copy()

missing_core_path = REGISTRY_DIR / "stage0_missing_core_files.csv"
missing_core_files.to_csv(missing_core_path, index=False)

display(download_summary)

print(f"Downloaded files: {download_summary['exists'].sum()} / {len(download_summary)}")
print(f"Missing core files: {len(missing_core_files)}")

if len(missing_core_files) > 0:
    display(missing_core_files)
else:
    print("All core NHANES files are present.")
```

```python
def test_read_xpt_files(raw_data_dir, max_files=8):
    records = []

    paths = sorted(Path(raw_data_dir).glob("*.XPT"))[:max_files]

    for path in paths:
        try:
            df = pd.read_sas(
                path,
                format="xport",
                encoding="utf-8"
            )

            records.append({
                "filename": path.name,
                "status": "read_ok",
                "rows": int(df.shape[0]),
                "columns": int(df.shape[1]),
                "sample_columns": ", ".join(df.columns[:8].astype(str).tolist())
            })

        except Exception as error:
            records.append({
                "filename": path.name,
                "status": "read_failed",
                "rows": 0,
                "columns": 0,
                "sample_columns": "",
                "error": str(error)
            })

    return pd.DataFrame(records)

xpt_read_test = test_read_xpt_files(
    RAW_DATA_DIR,
    max_files=12
)

read_test_path = REGISTRY_DIR / "stage0_xpt_read_test.csv"
xpt_read_test.to_csv(read_test_path, index=False)

display(xpt_read_test)

print(f"Read test saved to: {read_test_path}")

if (xpt_read_test["status"] == "read_ok").all():
    print("Read test passed.")
else:
    print("Some files failed the read test. Send me the failed rows.")
```

```python
import time
import hashlib
import requests
from tqdm.auto import tqdm

RAW_DATA_DIR = BASE_DIR / "data" / "raw_nhanes"

def file_sha256(path, chunk_size=1024 * 1024):
    hasher = hashlib.sha256()

    with open(path, "rb") as f:
        while True:
            chunk = f.read(chunk_size)

            if not chunk:
                break

            hasher.update(chunk)

    return hasher.hexdigest()


def download_file(url, destination, min_bytes=1024, retries=5, timeout=120):
    destination = Path(destination)
    destination.parent.mkdir(parents=True, exist_ok=True)

    if destination.exists() and destination.stat().st_size >= min_bytes:
        return {
            "status": "exists",
            "path": str(destination),
            "bytes": int(destination.stat().st_size),
            "sha256": file_sha256(destination)
        }

    temp_path = destination.with_suffix(destination.suffix + ".part")
    last_error = None

    for attempt in range(1, retries + 1):
        try:
            print(f"Attempt {attempt}: {url}")

            with requests.get(url, stream=True, timeout=timeout) as response:
                response.raise_for_status()

                total = int(response.headers.get("content-length", 0))

                with open(temp_path, "wb") as f:
                    with tqdm(
                        total=total,
                        unit="B",
                        unit_scale=True,
                        desc=destination.name,
                        leave=False
                    ) as progress:
                        for chunk in response.iter_content(chunk_size=1024 * 1024):
                            if chunk:
                                f.write(chunk)
                                progress.update(len(chunk))

            if temp_path.stat().st_size < min_bytes:
                raise ValueError(f"Downloaded file too small: {temp_path.stat().st_size} bytes")

            temp_path.replace(destination)

            return {
                "status": "downloaded",
                "path": str(destination),
                "bytes": int(destination.stat().st_size),
                "sha256": file_sha256(destination)
            }

        except Exception as error:
            last_error = str(error)

            if temp_path.exists():
                temp_path.unlink()

            time.sleep(5)

    return {
        "status": "failed",
        "path": str(destination),
        "bytes": 0,
        "sha256": None,
        "error": last_error
    }


retry_records = []

for row in failed.itertuples(index=False):
    destination = RAW_DATA_DIR / row.filename

    result = download_file(
        url=row.url,
        destination=destination,
        min_bytes=2048,
        retries=5,
        timeout=120
    )

    retry_records.append({
        "cycle": row.cycle,
        "module": row.module,
        "filename": row.filename,
        "url": row.url,
        **result
    })

retry_df = pd.DataFrame(retry_records)

retry_path = REGISTRY_DIR / "stage0_failed_file_retry_registry.csv"
retry_df.to_csv(retry_path, index=False)

display(retry_df)

print(f"Retry registry saved to: {retry_path}")
```

```python
import json
import pandas as pd
from pathlib import Path
import time

BASE_DIR = Path.home() / "projects" / "diabetes-defer-lab-tests"
REGISTRY_DIR = BASE_DIR / "download_registry"

summary_path = REGISTRY_DIR / "stage0_download_summary.csv"
manifest_path = REGISTRY_DIR / "stage0_manifest.json"
known_missing_path = REGISTRY_DIR / "stage0_known_acceptable_missing_files.csv"

download_summary = pd.read_csv(summary_path)

known_acceptable_missing = pd.DataFrame([
    {
        "cycle": "2017-2018",
        "module": "OGTT",
        "filename": "OGTT_J.XPT",
        "reason": (
            "OGTT_J.XPT was not available from the expected 2017-2018 CDC NHANES "
            "DataFiles URL. This is acceptable because the primary project target "
            "uses HbA1c and fasting glucose, while OGTT is treated as an optional "
            "secondary lab feature."
        ),
        "decision": "continue_without_file"
    }
])

known_acceptable_missing.to_csv(known_missing_path, index=False)

download_summary["known_acceptable_missing"] = False

mask = (
    (download_summary["cycle"] == "2017-2018")
    & (download_summary["module"] == "OGTT")
    & (download_summary["filename"] == "OGTT_J.XPT")
    & (~download_summary["exists"])
)

download_summary.loc[mask, "known_acceptable_missing"] = True

effective_missing_core = download_summary[
    (download_summary["is_core"])
    & (~download_summary["exists"])
    & (~download_summary["known_acceptable_missing"])
].copy()

download_summary.to_csv(summary_path, index=False)
effective_missing_core.to_csv(REGISTRY_DIR / "stage0_effective_missing_core_files.csv", index=False)

with open(manifest_path, "r") as f:
    manifest = json.load(f)

manifest["updated_at"] = time.strftime("%Y-%m-%d %H:%M:%S")
manifest["known_acceptable_missing_files_path"] = str(known_missing_path)
manifest["effective_missing_core_files"] = int(len(effective_missing_core))
manifest["stage0_decision"] = (
    "Proceed. OGTT_J.XPT is treated as acceptable missing. "
    "HbA1c and fasting glucose remain available for the main diabetes screening target."
)

with open(manifest_path, "w") as f:
    json.dump(manifest, f, indent=2)

print("Known acceptable missing files:")
display(known_acceptable_missing)

print("Effective missing core files:", len(effective_missing_core))
display(effective_missing_core)

print("Updated manifest:")
print(json.dumps(manifest, indent=2))
```

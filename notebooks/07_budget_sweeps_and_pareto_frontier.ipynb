```python
def save_plotly(
    fig,
    name,
    width=1200,
    height=750
):
    html_path = (
        FIGURE_DIR
        / f"{name}.html"
    )

    png_path = (
        FIGURE_DIR
        / f"{name}.png"
    )

    fig.write_html(
        html_path
    )

    try:
        fig.write_image(
            png_path,
            width=width,
            height=height,
            scale=2
        )
    except Exception as error:
        print(
            f"PNG export skipped for "
            f"{name}: {error}"
        )

        png_path = None

    return {
        "html": str(html_path),
        "png": (
            str(png_path)
            if png_path is not None
            else None
        )
    }


def save_json(
    obj,
    path
):
    path = Path(path)

    path.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    with open(
        path,
        "w"
    ) as f:
        json.dump(
            obj,
            f,
            indent=2,
            default=str
        )

    return path
```

```python
class MLPBlock(nn.Module):

    def __init__(
        self,
        input_dim,
        hidden_dim=128,
        num_layers=3,
        dropout=0.15
    ):
        super().__init__()

        layers = []

        current_dim = input_dim

        for _ in range(
            num_layers
        ):
            layers.extend([
                nn.Linear(
                    current_dim,
                    hidden_dim
                ),

                nn.LayerNorm(
                    hidden_dim
                ),

                nn.GELU(),

                nn.Dropout(
                    dropout
                )
            ])

            current_dim = (
                hidden_dim
            )

        self.network = (
            nn.Sequential(
                *layers
            )
        )

    def forward(
        self,
        x
    ):
        return self.network(
            x
        )


class DeferToLabNet(nn.Module):

    def __init__(
        self,
        cheap_input_dim,
        full_input_dim,
        hidden_dim=128,
        num_layers=3,
        dropout=0.15
    ):
        super().__init__()

        self.cheap_encoder = MLPBlock(
            cheap_input_dim,
            hidden_dim,
            num_layers,
            dropout
        )

        self.full_encoder = MLPBlock(
            full_input_dim,
            hidden_dim,
            num_layers,
            dropout
        )

        self.gate_encoder = MLPBlock(
            cheap_input_dim,
            hidden_dim,
            2,
            dropout
        )

        self.cheap_head = nn.Linear(
            hidden_dim,
            1
        )

        self.full_head = nn.Linear(
            hidden_dim,
            1
        )

        self.defer_head = nn.Linear(
            hidden_dim,
            1
        )

    def forward(
        self,
        x_cheap,
        x_full
    ):
        cheap_rep = (
            self.cheap_encoder(
                x_cheap
            )
        )

        full_rep = (
            self.full_encoder(
                x_full
            )
        )

        gate_rep = (
            self.gate_encoder(
                x_cheap
            )
        )

        cheap_logit = (
            self.cheap_head(
                cheap_rep
            )
            .squeeze(-1)
        )

        full_logit = (
            self.full_head(
                full_rep
            )
            .squeeze(-1)
        )

        defer_logit = (
            self.defer_head(
                gate_rep
            )
            .squeeze(-1)
        )

        return {
            "cheap_logit": (
                cheap_logit
            ),

            "full_logit": (
                full_logit
            ),

            "defer_probability": (
                torch.sigmoid(
                    defer_logit
                )
            )
        }
```

```python
@torch.no_grad()
def neural_predict(
    model,
    X_cheap,
    X_full,
    batch_size=2048
):
    cheap_probs = []
    full_probs = []
    defer_probs = []

    for start in range(
        0,
        len(X_cheap),
        batch_size
    ):
        end = (
            start
            + batch_size
        )

        xc = torch.from_numpy(
            X_cheap[
                start:end
            ]
        ).to(
            DEVICE
        )

        xf = torch.from_numpy(
            X_full[
                start:end
            ]
        ).to(
            DEVICE
        )

        outputs = model(
            xc,
            xf
        )

        cheap_probs.append(
            torch.sigmoid(
                outputs[
                    "cheap_logit"
                ]
            )
            .float()
            .cpu()
            .numpy()
        )

        full_probs.append(
            torch.sigmoid(
                outputs[
                    "full_logit"
                ]
            )
            .float()
            .cpu()
            .numpy()
        )

        defer_probs.append(
            outputs[
                "defer_probability"
            ]
            .float()
            .cpu()
            .numpy()
        )

    return {
        "cheap_probability": (
            np.concatenate(
                cheap_probs
            ).astype(
                np.float32
            )
        ),

        "full_probability": (
            np.concatenate(
                full_probs
            ).astype(
                np.float32
            )
        ),

        "defer_probability": (
            np.concatenate(
                defer_probs
            ).astype(
                np.float32
            )
        )
    }


val_outputs = neural_predict(
    model,
    X_val_cheap,
    X_val_full
)

test_outputs = neural_predict(
    model,
    X_test_cheap,
    X_test_full
)

print(
    "Validation predictions:",
    len(
        val_outputs[
            "defer_probability"
        ]
    )
)

print(
    "Test predictions:",
    len(
        test_outputs[
            "defer_probability"
        ]
    )
)
```

```python
def laboratory_rule(
    df
):
    hba1c = pd.to_numeric(
        df["hba1c"],
        errors="coerce"
    )

    glucose = pd.to_numeric(
        df["fasting_glucose"],
        errors="coerce"
    )

    return (
        (hba1c >= 6.5)
        |
        (glucose >= 126)
    ).astype(
        int
    ).to_numpy()


lab_rule_val = laboratory_rule(
    val_df
)

lab_rule_test = laboratory_rule(
    test_df
)

print(
    "Validation lab-rule agreement:",
    accuracy_score(
        y_val,
        lab_rule_val
    )
)

print(
    "Test lab-rule agreement:",
    accuracy_score(
        y_test,
        lab_rule_test
    )
)
```

```python
def find_youden_threshold(
    y_true,
    probability
):
    fpr, tpr, thresholds = (
        roc_curve(
            y_true,
            probability
        )
    )

    idx = np.nanargmax(
        tpr - fpr
    )

    return float(
        thresholds[idx]
    )


if (
    "cheap_classification_threshold"
    in checkpoint
):
    CHEAP_THRESHOLD = float(
        checkpoint[
            "cheap_classification_threshold"
        ]
    )
else:
    CHEAP_THRESHOLD = (
        find_youden_threshold(
            y_val,
            val_outputs[
                "cheap_probability"
            ]
        )
    )

print(
    "Cheap classification threshold:",
    round(
        CHEAP_THRESHOLD,
        5
    )
)
```

```python
def threshold_for_budget(
    validation_scores,
    budget
):
    scores = np.asarray(
        validation_scores,
        dtype=np.float32
    )

    budget = float(
        np.clip(
            budget,
            0.0,
            1.0
        )
    )

    if budget <= 0:
        return np.inf

    if budget >= 1:
        return -np.inf

    quantile = (
        1.0
        - budget
    )

    try:
        threshold = np.quantile(
            scores,
            quantile,
            method="linear"
        )
    except TypeError:
        threshold = np.quantile(
            scores,
            quantile,
            interpolation="linear"
        )

    return float(
        threshold
    )
```

```python
LAB_COST = 1.0

FALSE_NEGATIVE_COST = 10.0

FALSE_POSITIVE_COST = 2.0


def evaluate_referral_policy(
    y_true,
    cheap_probability,
    cheap_threshold,
    referral_mask,
    laboratory_prediction
):
    y_true = np.asarray(
        y_true
    ).astype(int)

    cheap_probability = np.asarray(
        cheap_probability
    )

    referral_mask = np.asarray(
        referral_mask
    ).astype(bool)

    laboratory_prediction = np.asarray(
        laboratory_prediction
    ).astype(int)

    cheap_prediction = (
        cheap_probability
        >= cheap_threshold
    ).astype(int)

    final_prediction = (
        cheap_prediction.copy()
    )

    final_prediction[
        referral_mask
    ] = laboratory_prediction[
        referral_mask
    ]

    tn, fp, fn, tp = (
        confusion_matrix(
            y_true,
            final_prediction,
            labels=[0, 1]
        )
        .ravel()
    )

    sensitivity = (
        tp / (tp + fn)
        if (tp + fn) > 0
        else np.nan
    )

    specificity = (
        tn / (tn + fp)
        if (tn + fp) > 0
        else np.nan
    )

    referral_n = int(
        referral_mask.sum()
    )

    referral_rate = float(
        referral_mask.mean()
    )

    positives = int(
        (y_true == 1).sum()
    )

    missed_case_rate = (
        fn / positives
        if positives > 0
        else np.nan
    )

    referral_positive_rate = (
        y_true[
            referral_mask
        ].mean()
        if referral_n > 0
        else np.nan
    )

    recovered_false_negatives = int(
        (
            referral_mask
            &
            (cheap_prediction == 0)
            &
            (y_true == 1)
            &
            (
                laboratory_prediction
                == 1
            )
        ).sum()
    )

    corrected_false_positives = int(
        (
            referral_mask
            &
            (cheap_prediction == 1)
            &
            (y_true == 0)
            &
            (
                laboratory_prediction
                == 0
            )
        ).sum()
    )

    tests_per_recovered_fn = (
        referral_n
        / recovered_false_negatives
        if recovered_false_negatives > 0
        else np.nan
    )

    average_lab_cost = (
        LAB_COST
        * referral_rate
    )

    false_negative_cost = (
        FALSE_NEGATIVE_COST
        * fn
        / len(y_true)
    )

    false_positive_cost = (
        FALSE_POSITIVE_COST
        * fp
        / len(y_true)
    )

    total_cost = (
        average_lab_cost
        + false_negative_cost
        + false_positive_cost
    )

    return {
        "n": int(
            len(y_true)
        ),

        "referral_n": (
            referral_n
        ),

        "referral_rate": (
            referral_rate
        ),

        "accuracy": (
            accuracy_score(
                y_true,
                final_prediction
            )
        ),

        "balanced_accuracy": (
            balanced_accuracy_score(
                y_true,
                final_prediction
            )
        ),

        "precision": (
            precision_score(
                y_true,
                final_prediction,
                zero_division=0
            )
        ),

        "sensitivity": (
            sensitivity
        ),

        "specificity": (
            specificity
        ),

        "f1": (
            f1_score(
                y_true,
                final_prediction,
                zero_division=0
            )
        ),

        "tp": int(tp),
        "tn": int(tn),
        "fp": int(fp),
        "fn": int(fn),

        "missed_case_rate": (
            missed_case_rate
        ),

        "referral_positive_rate": (
            referral_positive_rate
        ),

        "recovered_false_negatives": (
            recovered_false_negatives
        ),

        "corrected_false_positives": (
            corrected_false_positives
        ),

        "tests_per_recovered_fn": (
            tests_per_recovered_fn
        ),

        "average_lab_cost": (
            average_lab_cost
        ),

        "false_negative_cost": (
            false_negative_cost
        ),

        "false_positive_cost": (
            false_positive_cost
        ),

        "total_normalized_cost": (
            total_cost
        )
    }
```

```python
BUDGET_GRID = np.round(
    np.arange(
        0.0,
        1.0001,
        0.05
    ),
    2
)

print(
    BUDGET_GRID
)

print(
    "Operating points:",
    len(
        BUDGET_GRID
    )
)
```

```python
DEPLOYABLE_POLICIES = {
    "learned_defer_gate": (
        val_learned_score,
        test_learned_score
    ),

    "cheap_risk_referral": (
        val_risk_score,
        test_risk_score
    ),

    "uncertainty_referral": (
        val_uncertainty_score,
        test_uncertainty_score
    )
}

budget_records = []

for policy_name, (
    validation_scores,
    test_scores
) in DEPLOYABLE_POLICIES.items():

    for requested_budget in BUDGET_GRID:

        threshold = (
            threshold_for_budget(
                validation_scores,
                requested_budget
            )
        )

        referral_mask = (
            test_scores
            >= threshold
        )

        metrics = (
            evaluate_referral_policy(
                y_true=y_test,
                cheap_probability=(
                    test_outputs[
                        "cheap_probability"
                    ]
                ),
                cheap_threshold=(
                    CHEAP_THRESHOLD
                ),
                referral_mask=(
                    referral_mask
                ),
                laboratory_prediction=(
                    lab_rule_test
                )
            )
        )

        budget_records.append({
            "policy": (
                policy_name
            ),

            "requested_budget": (
                float(
                    requested_budget
                )
            ),

            "validation_threshold": (
                float(
                    threshold
                )
            ),

            **metrics
        })


budget_sweep = pd.DataFrame(
    budget_records
)

display(
    budget_sweep.head()
)
```

```python
RANDOM_REPEATS = 100

random_records = []

rng = np.random.default_rng(
    42
)

n_test = len(
    y_test
)

for requested_budget in BUDGET_GRID:

    k = int(
        round(
            requested_budget
            * n_test
        )
    )

    for repeat in range(
        RANDOM_REPEATS
    ):

        referral_mask = (
            np.zeros(
                n_test,
                dtype=bool
            )
        )

        if k > 0:

            chosen = rng.choice(
                n_test,
                size=min(
                    k,
                    n_test
                ),
                replace=False
            )

            referral_mask[
                chosen
            ] = True

        metrics = (
            evaluate_referral_policy(
                y_true=y_test,
                cheap_probability=(
                    test_outputs[
                        "cheap_probability"
                    ]
                ),
                cheap_threshold=(
                    CHEAP_THRESHOLD
                ),
                referral_mask=(
                    referral_mask
                ),
                laboratory_prediction=(
                    lab_rule_test
                )
            )
        )

        random_records.append({
            "policy": (
                "random_referral"
            ),

            "requested_budget": (
                float(
                    requested_budget
                )
            ),

            "repeat": (
                repeat
            ),

            **metrics
        })


random_results = pd.DataFrame(
    random_records
)

random_summary = (
    random_results
    .groupby(
        "requested_budget",
        observed=True
    )
    .agg(
        referral_rate=(
            "referral_rate",
            "mean"
        ),

        sensitivity=(
            "sensitivity",
            "mean"
        ),

        specificity=(
            "specificity",
            "mean"
        ),

        balanced_accuracy=(
            "balanced_accuracy",
            "mean"
        ),

        missed_case_rate=(
            "missed_case_rate",
            "mean"
        ),

        total_normalized_cost=(
            "total_normalized_cost",
            "mean"
        ),

        recovered_false_negatives=(
            "recovered_false_negatives",
            "mean"
        ),

        sensitivity_low=(
            "sensitivity",
            lambda x: np.quantile(
                x,
                0.025
            )
        ),

        sensitivity_high=(
            "sensitivity",
            lambda x: np.quantile(
                x,
                0.975
            )
        ),

        cost_low=(
            "total_normalized_cost",
            lambda x: np.quantile(
                x,
                0.025
            )
        ),

        cost_high=(
            "total_normalized_cost",
            lambda x: np.quantile(
                x,
                0.975
            )
        )
    )
    .reset_index()
)

random_summary[
    "policy"
] = "random_referral"

display(
    random_summary.head()
)
```

```python
cheap_test_prediction = (
    test_outputs[
        "cheap_probability"
    ]
    >= CHEAP_THRESHOLD
).astype(int)

oracle_error_cost = np.zeros(
    len(y_test),
    dtype=np.float32
)

oracle_error_cost[
    (
        cheap_test_prediction == 0
    )
    &
    (
        y_test == 1
    )
] = (
    FALSE_NEGATIVE_COST
)

oracle_error_cost[
    (
        cheap_test_prediction == 1
    )
    &
    (
        y_test == 0
    )
] = (
    FALSE_POSITIVE_COST
)

oracle_order = np.argsort(
    -oracle_error_cost
)

oracle_records = []

for requested_budget in BUDGET_GRID:

    k = int(
        round(
            requested_budget
            * len(y_test)
        )
    )

    referral_mask = np.zeros(
        len(y_test),
        dtype=bool
    )

    if k > 0:
        referral_mask[
            oracle_order[:k]
        ] = True

    metrics = (
        evaluate_referral_policy(
            y_true=y_test,
            cheap_probability=(
                test_outputs[
                    "cheap_probability"
                ]
            ),
            cheap_threshold=(
                CHEAP_THRESHOLD
            ),
            referral_mask=(
                referral_mask
            ),
            laboratory_prediction=(
                lab_rule_test
            )
        )
    )

    oracle_records.append({
        "policy": (
            "oracle_upper_bound"
        ),

        "requested_budget": (
            float(
                requested_budget
            )
        ),

        **metrics
    })


oracle_sweep = pd.DataFrame(
    oracle_records
)

display(
    oracle_sweep.head()
)
```

```python
def mark_pareto_frontier(
    df,
    x_col="referral_rate",
    y_col="sensitivity"
):
    work = df.copy()

    is_pareto = np.ones(
        len(work),
        dtype=bool
    )

    x = work[
        x_col
    ].to_numpy()

    y = work[
        y_col
    ].to_numpy()

    for i in range(
        len(work)
    ):

        dominated = (
            (x <= x[i])
            &
            (y >= y[i])
            &
            (
                (x < x[i])
                |
                (y > y[i])
            )
        )

        dominated[i] = False

        if dominated.any():
            is_pareto[i] = False

    work[
        "pareto_efficient"
    ] = is_pareto

    return work


pareto_candidates = (
    budget_sweep.copy()
)

pareto_results = (
    mark_pareto_frontier(
        pareto_candidates
    )
)

pareto_frontier = (
    pareto_results[
        pareto_results[
            "pareto_efficient"
        ]
    ]
    .sort_values(
        "referral_rate"
    )
)

display(
    pareto_frontier[
        [
            "policy",
            "requested_budget",
            "referral_rate",
            "sensitivity",
            "specificity",
            "balanced_accuracy",
            "total_normalized_cost"
        ]
    ]
)

pareto_frontier.to_csv(
    TABLE_DIR
    / "step11_pareto_frontier.csv",
    index=False
)
```

```python
def get_device(prefer_gpu: bool = True) -> torch.device:
    if prefer_gpu and torch.cuda.is_available():
        return torch.device("cuda")
    return torch.device("cpu")


DEVICE = get_device(prefer_gpu=True)
USE_AMP = DEVICE.type == "cuda"


def seed_everything(seed: int = 42) -> None:
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)

    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


GLOBAL_SEED = 42
seed_everything(GLOBAL_SEED)

print("Device:", DEVICE)
print("Mixed precision enabled:", USE_AMP)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
    print("CUDA version:", torch.version.cuda)
    print("GPU memory allocated:", round(torch.cuda.memory_allocated() / 1024**3, 3), "GB")
    print("GPU memory reserved:", round(torch.cuda.memory_reserved() / 1024**3, 3), "GB")
else:
    print("CUDA is not available. Neural sections will run on CPU.")
```

```python
@dataclass
class DataConfig:
    raw_data_dir: str = str(RAW_DATA_DIR)
    interim_data_dir: str = str(INTERIM_DATA_DIR)
    processed_data_dir: str = str(PROCESSED_DATA_DIR)
    id_col: str = "SEQN"
    cycle_col: str = "cycle"
    target_col: str = "diabetes_binary"
    secondary_target_col: str = "dysglycemia_binary"
    validation_size: float = 0.15
    test_size: float = 0.20
    random_state: int = GLOBAL_SEED
    use_cycle_split: bool = True
    min_age: int = 18
    save_processed_files: bool = True


@dataclass
class CostConfig:
    cheap_feature_cost: float = 0.0
    hba1c_cost: float = 1.0
    fasting_glucose_cost: float = 0.8
    ogtt_cost: float = 1.5
    false_negative_cost: float = 10.0
    false_positive_cost: float = 2.0
    default_budget_per_patient: float = 0.35
    budget_grid: Tuple[float, ...] = (
        0.0, 0.05, 0.10, 0.20, 0.35, 0.50, 0.75, 1.00
    )


@dataclass
class ModelConfig:
    random_state: int = GLOBAL_SEED
    batch_size: int = 512
    num_epochs: int = 80
    learning_rate: float = 1e-3
    weight_decay: float = 1e-4
    hidden_dim: int = 128
    num_hidden_layers: int = 3
    dropout: float = 0.15
    early_stopping_patience: int = 10
    min_delta: float = 1e-4
    defer_cost_lambda: float = 0.10
    budget_penalty_lambda: float = 10.0
    use_amp: bool = USE_AMP
    device: str = str(DEVICE)


@dataclass
class EvaluationConfig:
    threshold_grid_size: int = 201
    default_threshold: float = 0.50
    sensitivity_targets: Tuple[float, ...] = (0.80, 0.85, 0.90, 0.95)
    calibration_bins: int = 10
    bootstrap_iterations: int = 500
    subgroup_columns: Tuple[str, ...] = (
        "sex",
        "age_group",
        "bmi_group",
        "race_ethnicity",
        "cycle",
        "uspstf_age_bmi_eligible",
        "overweight_or_obesity",
        "obesity",
        "hypertension_range_bp"
    )


@dataclass
class ProjectConfig:
    project_name: str = "Learning to Defer to Lab Tests"
    project_slug: str = "diabetes-defer-lab-tests"
    base_dir: str = str(BASE_DIR)
    output_dir: str = str(OUTPUT_DIR)
    figure_dir: str = str(FIGURE_DIR)
    table_dir: str = str(TABLE_DIR)
    model_dir: str = str(MODEL_DIR)
    report_dir: str = str(REPORT_DIR)
    data: DataConfig = DataConfig()
    cost: CostConfig = CostConfig()
    model: ModelConfig = ModelConfig()
    evaluation: EvaluationConfig = EvaluationConfig()


CONFIG = ProjectConfig()

print(json.dumps(asdict(CONFIG), indent=2, default=str))
```

```python
def timestamp() -> str:
    return time.strftime("%Y%m%d_%H%M%S")


def save_json(obj: Dict[str, Any], path: Path) -> Path:
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)

    with open(path, "w") as f:
        json.dump(obj, f, indent=2, default=str)

    return path


def save_table(
    df: pd.DataFrame,
    filename: str,
    index: bool = False
) -> Path:
    path = TABLE_DIR / filename
    path.parent.mkdir(parents=True, exist_ok=True)

    if filename.endswith(".csv"):
        df.to_csv(path, index=index)
    elif filename.endswith(".parquet"):
        df.to_parquet(path, index=index)
    else:
        df.to_csv(path.with_suffix(".csv"), index=index)
        path = path.with_suffix(".csv")

    return path


def save_matplotlib_figure(
    fig,
    filename: str,
    dpi: int = 300,
    bbox_inches: str = "tight"
) -> Path:
    path = FIGURE_DIR / filename

    if path.suffix == "":
        path = path.with_suffix(".png")

    path.parent.mkdir(parents=True, exist_ok=True)
    fig.savefig(path, dpi=dpi, bbox_inches=bbox_inches)
    return path


def save_plotly_figure(
    fig,
    filename: str,
    width: int = 1200,
    height: int = 800
) -> Dict[str, Path]:
    html_path = FIGURE_DIR / f"{filename}.html"
    png_path = FIGURE_DIR / f"{filename}.png"

    html_path.parent.mkdir(parents=True, exist_ok=True)

    fig.write_html(html_path)

    try:
        fig.write_image(png_path, width=width, height=height, scale=2)
    except Exception as error:
        print(f"PNG export skipped for {filename}: {error}")
        png_path = None

    return {
        "html": html_path,
        "png": png_path
    }


def clean_column_names(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    df.columns = [str(col).strip() for col in df.columns]
    return df


def memory_usage_report(df: pd.DataFrame, name: str = "dataframe") -> None:
    memory_mb = df.memory_usage(deep=True).sum() / 1024**2
    print(f"{name}: {df.shape[0]:,} rows x {df.shape[1]:,} columns, {memory_mb:.2f} MB")


def reduce_dataframe_memory(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()

    for column in df.select_dtypes(include=["float64"]).columns:
        df[column] = pd.to_numeric(df[column], downcast="float")

    for column in df.select_dtypes(include=["int64"]).columns:
        df[column] = pd.to_numeric(df[column], downcast="integer")

    return df


def clear_memory() -> None:
    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()


print("Utility functions ready.")
```

```python
def plot_palette_preview() -> None:
    labels = list(PROJECT_COLORS.keys())
    colors = list(PROJECT_COLORS.values())

    fig, ax = plt.subplots(figsize=(12, 5))

    for idx, (label, color) in enumerate(zip(labels, colors)):
        ax.bar(idx, 1, color=color)
        ax.text(
            idx,
            -0.08,
            label,
            ha="right",
            va="top",
            rotation=45,
            fontsize=9
        )

    ax.set_xlim(-0.7, len(labels) - 0.3)
    ax.set_ylim(-0.35, 1.05)
    ax.set_yticks([])
    ax.set_xticks([])
    ax.set_title("Project Color System")
    ax.set_frame_on(False)

    plt.show()

    save_matplotlib_figure(fig, "step1_project_palette_preview.png")


plot_palette_preview()
```
